```polyglot-notebook
public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;

    public Product(string name, decimal price, int quantity, string category, string description)
    {
        this.name = name;
        this.price = price;
        this.quantity = quantity;
        this.category = category;
        this.description = description;
    }

    public decimal GetTotalPrice()
    {
        return price * quantity;
    }

    public void UpdateQuantity(int amount)
    {
        quantity += amount;
    }

    public string GetProductionInfo()
    {
        return $"Имя продукта: {name}, Цена: {price}, Количество: {quantity}, Категория: {category}, Описание: {description}";
    }
}

public class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contractinfo;
    private List<Product> products = new List<Product>();

    public Seller(string name, string employeeId, decimal salary, string contractinfo)
    {
        this.name = name;
        this.employeeId = employeeId;
        this.salary = salary;
        this.contractinfo = contractinfo;
    }

    public void AddProduct(Product product)
    {
        products.Add(product);
    }

    public void SellProduct(Product product, int quantity)
    {
        product.UpdateQuantity(-quantity);
    }

    public string GetSellerInfo()
    {
        return $"Имя: {name}, ID сотрудника: {employeeId}, Зарплата: {salary}, Информация о контракте: {contractinfo}";
    }
}

public class Store
{
    private string storeName;
    private string location;
    public string storeHours;
    private List<Seller> sellers = new List<Seller>();
    private List<Product> products = new List<Product>();

    public Store(string storeName, string location, string storeHours)
    {
        this.storeName = storeName;
        this.location = location;
        this.storeHours = storeHours;
    }

    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);
    }

    public void ListProducts()
    {
        foreach (var product in products)
        {
            Console.WriteLine(product.GetProductionInfo());
        }
    }

    public string GetStoreInfo()
    {
        return $"Имя магазина: {storeName}, Где находится: {location}, Часы работы: {storeHours}";
    }
}
Product GLaptop = new Product("Ноутбук", 69.999m, 10, "Электроника", "Игровой ноутбук");
Product GMouse = new Product("Мышь", 12.999m, 15, "Электроника", "Игровая мышь Razer");
Seller seller1 = new Seller("Дмитрий Ицков", "А001", 5000000, "Полная ставка");
Seller seller2 = new Seller("Виктор Яковлев", "А002", 2500, "Полставки");
Store TStore1 = new Store("Tech Store", "Мельникайте", "9:00-21:00");
Store TStore2 = new Store("DNS", "Харьковская", "11:00-22:00");
Console.WriteLine(GLaptop.GetProductionInfo());
Console.WriteLine(GMouse.GetProductionInfo());
Console.WriteLine($"Общая стоимость товара Игровой ноутбук: {GLaptop.GetTotalPrice()}");
Console.WriteLine($"Общая стоимость товара Игровая мышь Razer: {GMouse.GetTotalPrice()}");
seller1.AddProduct(GLaptop);
seller2.AddProduct(GMouse);
seller1.SellProduct(GLaptop, 2);
Console.WriteLine($"В магазине Tech Store продано 2 игровых ноутбука, общая стоимость данного товара в данном магазине: {GLaptop.GetTotalPrice()}");
TStore1.AddSeller(seller1);
TStore2.AddSeller(seller2);
TStore1.ListProducts();
TStore2.ListProducts();
Console.WriteLine(seller1.GetSellerInfo());
Console.WriteLine(seller2.GetSellerInfo());
Console.WriteLine(TStore1.GetStoreInfo());
Console.WriteLine(TStore2.GetStoreInfo());
```
