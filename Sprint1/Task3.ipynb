Создать базовый класс Notification в C#, который будет представлять уведомления
пользователям. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;
using System.Collections.Generic;

public class Notification
{
    private int _notificationId;
    private string _messageText;
    private string _type;
    private DateTime _createdDate;

    public int NotificationId
    {
        get { return _notificationId; }
        set { _notificationId = value > 0 ? value : throw new ArgumentException("ID должен быть положительным"); }
    }

    public string MessageText
    {
        get { return _messageText; }
        set { _messageText = !string.IsNullOrWhiteSpace(value) ? value : throw new ArgumentException("Текст сообщения не может быть пустым"); }
    }

    public string Type
    {
        get { return _type; }
        set { _type = !string.IsNullOrWhiteSpace(value) ? value : throw new ArgumentException("Тип уведомления не может быть пустым"); }
    }

    public DateTime CreatedDate
    {
        get { return _createdDate; }
        private set { _createdDate = value; }
    }

    public Notification(int id, string message, string type)
    {
        NotificationId = id;
        MessageText = message;
        Type = type;
        CreatedDate = DateTime.Now;
    }

    public virtual void DisplayNotification()
    {
        Console.WriteLine($"Уведомление [{NotificationId}]: {MessageText} (Создано: {CreatedDate:HH:mm:ss})");
    }

    public virtual void SendNotification()
    {
        Console.WriteLine($"Отправка уведомления типа {Type}");
    }

    public virtual string GetNotificationDetails()
    {
        return $"ID: {NotificationId}, Тип: {Type}, Сообщение: {MessageText}, Создано: {CreatedDate:yyyy-MM-dd HH:mm:ss}";
    }
}

public class EmailNotification : Notification
{
    private string _emailAddress;

    public string EmailAddress
    {
        get { return _emailAddress; }
        set { _emailAddress = IsValidEmail(value) ? value : throw new ArgumentException("Неверный формат email адреса"); }
    }

    public EmailNotification(int id, string message, string email)
        : base(id, message, "Email")
    {
        EmailAddress = email;
    }

    private bool IsValidEmail(string email)
    {
        return !string.IsNullOrWhiteSpace(email) && email.Contains("@") && email.Contains(".");
    }

    public override void SendNotification()
    {
        base.SendNotification();
        Console.WriteLine($"Отправлено на email: {EmailAddress}");
        Console.WriteLine($"Имитация отправки email на адрес {EmailAddress} с текстом: {MessageText}");
    }

    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Email: {EmailAddress}";
    }
}

public class SMSNotification : Notification
{
    private string _phoneNumber;

    public string PhoneNumber
    {
        get { return _phoneNumber; }
        set { _phoneNumber = IsValidPhoneNumber(value) ? value : throw new ArgumentException("Неверный формат номера телефона"); }
    }

    public SMSNotification(int id, string message, string phone)
        : base(id, message, "SMS")
    {
        PhoneNumber = phone;
    }

    private bool IsValidPhoneNumber(string phone)
    {
        return !string.IsNullOrWhiteSpace(phone) && phone.Length >= 10 && phone.StartsWith("+");
    }

    public override void SendNotification()
    {
        base.SendNotification();
        Console.WriteLine($"Отправлено SMS на номер: {PhoneNumber}");
        Console.WriteLine($"Имитация отправки SMS на номер {PhoneNumber} с текстом: {MessageText}");
    }

    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Телефон: {PhoneNumber}";
    }
}

public class PushNotification : Notification
{
    private string _platform;

    public string Platform
    {
        get { return _platform; }
        set { _platform = !string.IsNullOrWhiteSpace(value) ? value : "Unknown"; }
    }

    public PushNotification(int id, string message, string platform)
        : base(id, message, "Push")
    {
        Platform = platform;
    }

    public override void DisplayNotification()
    {
        Console.WriteLine($"[{Platform}] Push-уведомление: {MessageText}");
    }

    public override void SendNotification()
    {
        Console.WriteLine($"Отправка push-уведомления на платформу {Platform}");
        Console.WriteLine($"Имитация отправки push-уведомления на {Platform} с текстом: {MessageText}");
    }

    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Платформа: {Platform}";
    }
}
Console.WriteLine("Демонстрация системы уведомлений:");
Console.WriteLine("==================================");

        // Создание списка уведомлений
var notifications = new List<Notification>
{
    new EmailNotification(1, "Добро пожаловать в нашу систему!", "user@example.com"),
    new SMSNotification(2, "Ваш код подтверждения: 1234", "+79991234567"),
    new PushNotification(3, "У вас новое сообщение", "iOS"),
    new PushNotification(4, "Обновление приложения доступно", "Android"),
};

        // Обработка всех уведомлений через foreach
foreach (var notification in notifications)
{
    Console.WriteLine("\n--- Обработка уведомления ---");
    notification.DisplayNotification();
    notification.SendNotification();
    Console.WriteLine("Детали: " + notification.GetNotificationDetails());
}

        // Статистика по уведомлениям
Console.WriteLine("\n--- Статистика уведомлений ---");
var emailCount = 0;
var smsCount = 0;
var pushCount = 0;

foreach (var notification in notifications)
{
    if (notification is EmailNotification) emailCount++;
    else if (notification is SMSNotification) smsCount++;
    else if (notification is PushNotification) pushCount++;
}

Console.WriteLine($"Всего уведомлений: {notifications.Count}");
Console.WriteLine($"Email: {emailCount}");
Console.WriteLine($"SMS: {smsCount}");
Console.WriteLine($"Push: {pushCount}");

Console.WriteLine("\n==================================");
Console.WriteLine("Обработка всех уведомлений завершена!");
```
