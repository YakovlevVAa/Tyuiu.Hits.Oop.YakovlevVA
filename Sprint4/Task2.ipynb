### Интерфейсы в C#

**Интерфейс** — это контракт, который определяет набор методов и свойств, которые класс должен реализовать. Интерфейсы позволяют создать гибкую и расширяемую архитектуру, поскольку они отделяют определение поведения от его реализации.

Давайте подробно разберем понятия интерфейсов и управления зависимостями в C# на примере классов `Vehicle`, `Car` и `Motorcycle`.

#### Пример интерфейса

Предположим, мы имеем интерфейс `IVehicle`, который определяет поведение транспортных средств. Ниже приведен пример его определения:

```csharp
public interface IVehicle
{
    void Start();            // Метод для запуска транспортного средства
    void Stop();             // Метод для остановки транспортного средства
    int GetNumberOfWheels(); // Метод для получения количества колес
}
```

Интерфейс `IVehicle` требует, чтобы любой класс, реализующий его, обеспечил реализацию этих методов.

### Реализация интерфейса

Теперь создадим классы `Car` и `Motorcycle`, которые будут реализовывать этот интерфейс:

```csharp
public class Car : IVehicle
{
    public void Start()
    {
        Console.WriteLine("Машина запущена.");
    }

    public void Stop()
    {
        Console.WriteLine("Машина остановлена.");
    }

    public int GetNumberOfWheels()
    {
        return 4; // У машины 4 колеса
    }
}

public class Motorcycle : IVehicle
{
    public void Start()
    {
        Console.WriteLine("Мотоцикл запущен.");
    }

    public void Stop()
    {
        Console.WriteLine("Мотоцикл остановлен.");
    }

    public int GetNumberOfWheels()
    {
        return 2; // У мотоцикла 2 колеса
    }
}
```

Каждый класс (`Car` и `Motorcycle`) реализует все методы, определенные в интерфейсе `IVehicle`, что позволяет создать единый интерфейс для работы с разными типами транспортных средств.

### Управление зависимостями (Dependency Injection)

**Управление зависимостями** — это паттерн проектирования, который помогает управлять созданием зависимостей в классе. Этот подход позволяет внедрять (или "вводить") зависимости через конструктор, методы или свойства, вместо того чтобы создавать их внутри класса. Это делает код более чистым и облегчает его тестирование.

#### Пример управления зависимостями

Создадим класс `VehicleService`, который будет зависеть от интерфейса `IVehicle`:

```csharp
public class VehicleService
{
    private readonly IVehicle _vehicle;

    // Внедрение зависимости через конструктор
    public VehicleService(IVehicle vehicle)
    {
        _vehicle = vehicle; 
    }

    public void UseVehicle()
    {
        _vehicle.Start(); // Запускаем транспортное средство
        Console.WriteLine($"У этого транспортного средства {_vehicle.GetNumberOfWheels()} колеса.");
        _vehicle.Stop(); // Останавливаем транспортное средство
    }
}
```

В этом примере класс `VehicleService` принимает в своем конструкторе объект, реализующий интерфейс `IVehicle`. Это позволяет `VehicleService` работать с любым объектом, который реализует `IVehicle`, обеспечивая гибкость и уменьшая связывание между классами.

### Использование классов

Теперь давайте создадим объекты `Car` и `Motorcycle` и используем наш сервис:

```csharp
IVehicle myCar = new Car(); // Создаем объект машины
VehicleService carService = new VehicleService(myCar); // Передаем машину в сервис
carService.UseVehicle(); // Используем сервис с машиной

IVehicle myMotorcycle = new Motorcycle(); // Создаем объект мотоцикла
VehicleService motorcycleService = new VehicleService(myMotorcycle); // Передаем мотоцикл в сервис
motorcycleService.UseVehicle(); // Используем сервис с мотоциклом

```

### Результат выполнения

При запуске данной программы будет следующий вывод:

```
Машина запущена.
У этого транспортного средства 4 колеса.
Машина остановлена.
Мотоцикл запущен.
У этого транспортного средства 2 колеса.
Мотоцикл остановлен.
```

### Заключение

Использование интерфейсов и управления зависимостями позволяет создать более организованный и тестируемый код. Интерфейсы определяют стандартные поведения, а управление зависимостями снижает связность между компонентами. Это делает систему более гибкой и упрощает процесс её поддержки и расширения в будущем.

````

public interface IVehicle
{
    void Start();
    void Stop();
    int GetNumberOfWheels();
}

public abstract class Vehicle : IVehicle
{
    public abstract int GetNumberOfWheels(); 

    public virtual void Start()
    {
        Console.WriteLine("Транспортное средство запущено.");
    }

    public virtual void Stop()
    {
        Console.WriteLine("Транспортное средство остановлено.");
    }
}

public class Car : Vehicle
{
    public override int GetNumberOfWheels()
    {
        return 4;
    }

    public override void Start()
    {
        Console.WriteLine("Машина запущена.");
    }

    public override void Stop()
    {
        Console.WriteLine("Машина остановлена.");
    }
}

public class Motorcycle : Vehicle
{
    public override int GetNumberOfWheels()
    {
        return 2;
    }

    public override void Start()
    {
        Console.WriteLine("Мотоцикл запущен.");
    }

    public override void Stop()
    {
        Console.WriteLine("Мотоцикл остановлен.");
    }
}


public class VehicleService
{
    private readonly IVehicle _vehicle;

    public VehicleService(IVehicle vehicle)
    {
        _vehicle = vehicle;
    }

    public void UseVehicle()
    {
        _vehicle.Start();
        Console.WriteLine($"У этого транспортного средства {_vehicle.GetNumberOfWheels()} колеса.");
        _vehicle.Stop();
    }
}


        IVehicle car = new Car();
        IVehicle motorcycle = new Motorcycle();

        VehicleService carService = new VehicleService(car);
        VehicleService motorcycleService = new VehicleService(motorcycle);

        carService.UseVehicle();
        Console.WriteLine();
        motorcycleService.UseVehicle();
````

````

public abstract class Animal
{
    // Поля с защищенным доступом
    private string _name;
    private int _age;
    private double _weight;

    // Конструктор базового класса
    protected Animal(string name, int age, double weight)
    {
        _name = name;
        _age = age;
        _weight = weight;
    }

    // Свойства с геттерами и сеттерами
    public string Name
    {
        get { return _name; }
        set
        {
            if (!string.IsNullOrWhiteSpace(value))
                _name = value;
            else
                throw new ArgumentException("Имя не может быть пустым");
        }
    }

    public int Age
    {
        get { return _age; }
        set
        {
            if (value >= 0)
                _age = value;
            else
                throw new ArgumentException("Возраст не может быть отрицательным");
        }
    }

    public double Weight
    {
        get { return _weight; }
        set
        {
            if (value > 0)
                _weight = value;
            else
                throw new ArgumentException("Вес должен быть положительным");
        }
    }

    // Абстрактные методы - должны быть реализованы в производных классах
    public abstract void MakeSound();
    public abstract string GetAnimalInfo();

    // Виртуальный метод с реализацией по умолчанию
    public virtual void Sleep()
    {
        Console.WriteLine($"{Name} спит");
    }

    // Общий метод для всех животных
    public void Eat(string food)
    {
        Console.WriteLine($"{Name} ест {food}");
    }
}

// Интерфейс для домашних животных
public interface IPet
{
    void Play();
    string Owner { get; set; }
}

// Интерфейс для обучаемых животных
public interface ITrainable
{
    void Train(string command);
    bool KnowsCommand(string command);
}

// Производный класс Dog
public class Dog : Animal, IPet, ITrainable
{
    public string Breed { get; set; }
    public string Owner { get; set; }
    private List<string> _knownCommands;

    public Dog(string name, int age, double weight, string breed, string owner) 
        : base(name, age, weight)
    {
        Breed = breed;
        Owner = owner;
        _knownCommands = new List<string>();
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} гавкает: Гав-гав!");
    }

    public override string GetAnimalInfo()
    {
        return $"Собака: {Name}, Порода: {Breed}, Возраст: {Age} лет, Вес: {Weight} кг, Хозяин: {Owner}";
    }

    public void Play()
    {
        Console.WriteLine($"{Name} играет с мячиком");
    }

    public void Train(string command)
    {
        if (!_knownCommands.Contains(command))
        {
            _knownCommands.Add(command);
            Console.WriteLine($"{Name} выучил команду: {command}");
        }
    }

    public bool KnowsCommand(string command)
    {
        return _knownCommands.Contains(command);
    }

    // Переопределение виртуального метода
    public override void Sleep()
    {
        Console.WriteLine($"{Name} спит на своей подстилке");
    }
}

// Производный класс Cat
public class Cat : Animal, IPet
{
    public string Color { get; set; }
    public string Owner { get; set; }
    public bool IsIndoor { get; set; }

    public Cat(string name, int age, double weight, string color, string owner, bool isIndoor)
        : base(name, age, weight)
    {
        Color = color;
        Owner = owner;
        IsIndoor = isIndoor;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }

    public override string GetAnimalInfo()
    {
        string indoorStatus = IsIndoor ? "домашний" : "уличный";
        return $"Кот: {Name}, Цвет: {Color}, Возраст: {Age} лет, Вес: {Weight} кг, {indoorStatus}";
    }

    public void Play()
    {
        Console.WriteLine($"{Name} играет с клубком ниток");
    }

    // Специфический метод для кота
    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: Мрррр...");
    }
}

// Производный класс Bird
public class Bird : Animal, ITrainable
{
    public string Species { get; set; }
    public double Wingspan { get; set; }
    private List<string> _knownCommands;

    public Bird(string name, int age, double weight, string species, double wingspan)
        : base(name, age, weight)
    {
        Species = species;
        Wingspan = wingspan;
        _knownCommands = new List<string>();
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }

    public override string GetAnimalInfo()
    {
        return $"Птица: {Name}, Вид: {Species}, Размах крыльев: {Wingspan} см, Возраст: {Age} лет";
    }

    public void Train(string command)
    {
        if (!_knownCommands.Contains(command))
        {
            _knownCommands.Add(command);
            Console.WriteLine($"{Name} научился: {command}");
        }
    }

    public bool KnowsCommand(string command)
    {
        return _knownCommands.Contains(command);
    }

    // Специфический метод для птицы
    public void Fly()
    {
        Console.WriteLine($"{Name} летает с размахом крыльев {Wingspan} см");
    }
}

// Производный класс Fish
public class Fish : Animal
{
    public string WaterType { get; set; }
    public double Length { get; set; }

    public Fish(string name, int age, double weight, string waterType, double length)
        : base(name, age, weight)
    {
        WaterType = waterType;
        Length = length;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} пускает пузыри: Буль-буль!");
    }

    public override string GetAnimalInfo()
    {
        return $"Рыба: {Name}, Тип воды: {WaterType}, Длина: {Length} см, Возраст: {Age} лет";
    }

    // Переопределение виртуального метода
    public override void Sleep()
    {
        Console.WriteLine($"{Name} спит, плавая в воде");
    }

    // Специфический метод для рыбы
    public void Swim()
    {
        Console.WriteLine($"{Name} плавает в {WaterType} воде");
    }
}

// Интерфейс для сервиса управления животными
public interface IAnimalService
{
    void AddAnimal(Animal animal);
    void RemoveAnimal(string name);
    Animal FindAnimal(string name);
    void DisplayAllAnimals();
    void MakeAllSounds();
}

// Реализация сервиса животных с управлением зависимостями
public class AnimalService : IAnimalService
{
    private readonly List<Animal> _animals;

    public AnimalService()
    {
        _animals = new List<Animal>();
    }

    public void AddAnimal(Animal animal)
    {
        _animals.Add(animal);
        Console.WriteLine($"Добавлено животное: {animal.Name}");
    }

    public void RemoveAnimal(string name)
    {
        var animal = _animals.Find(a => a.Name.Equals(name, StringComparison.OrdinalIgnoreCase));
        if (animal != null)
        {
            _animals.Remove(animal);
            Console.WriteLine($"Удалено животное: {name}");
        }
        else
        {
            Console.WriteLine($"Животное с именем {name} не найдено");
        }
    }

    public Animal FindAnimal(string name)
    {
        return _animals.Find(a => a.Name.Equals(name, StringComparison.OrdinalIgnoreCase));
    }

    public void DisplayAllAnimals()
    {
        Console.WriteLine("\n=== Все животные ===");
        foreach (var animal in _animals)
        {
            Console.WriteLine(animal.GetAnimalInfo());
        }
    }

    public void MakeAllSounds()
    {
        Console.WriteLine("\n=== Звуки животных ===");
        foreach (var animal in _animals)
        {
            animal.MakeSound();
        }
    }
}

// Класс для управления зоопарком с внедрением зависимостей
public class ZooManager
{
    private readonly IAnimalService _animalService;

    // Внедрение зависимости через конструктор
    public ZooManager(IAnimalService animalService)
    {
        _animalService = animalService;
    }

    public void RunZoo()
    {
        Console.WriteLine("=== Управление зоопарком ===");
        _animalService.DisplayAllAnimals();
        _animalService.MakeAllSounds();
    }

    public void AddNewAnimal(Animal animal)
    {
        _animalService.AddAnimal(animal);
    }

    public void FindAndDisplayAnimal(string name)
    {
        var animal = _animalService.FindAnimal(name);
        if (animal != null)
        {
            Console.WriteLine($"Найдено: {animal.GetAnimalInfo()}");
        }
        else
        {
            Console.WriteLine($"Животное с именем '{name}' не найдено");
        }
    }
}


        // Создание сервиса животных
        IAnimalService animalService = new AnimalService();
        
        // Внедрение зависимости в менеджер зоопарка
        ZooManager zooManager = new ZooManager(animalService);

        // Создание различных животных
        Dog dog = new Dog("Барсик", 3, 12.5, "Лабрадор", "Иван Петров");
        Cat cat = new Cat("Мурка", 2, 4.2, "Рыжий", "Мария Сидорова", true);
        Bird bird = new Bird("Кеша", 1, 0.3, "Попугай", 25.5);
        Fish fish = new Fish("Немо", 1, 0.5, "морская", 8.2);

        // Добавление животных через менеджер
        zooManager.AddNewAnimal(dog);
        zooManager.AddNewAnimal(cat);
        zooManager.AddNewAnimal(bird);
        zooManager.AddNewAnimal(fish);

        // Дрессировка животных
        dog.Train("Сидеть");
        dog.Train("Лежать");
        bird.Train("Говорить 'Привет'");

        // Демонстрация работы зоопарка
        zooManager.RunZoo();

        Console.WriteLine("\n=== Дополнительные действия ===");
        
        // Поиск животных
        zooManager.FindAndDisplayAnimal("Мурка");
        zooManager.FindAndDisplayAnimal("Несуществующий");

        // Специфические действия животных
        dog.Play();
        cat.Purr();
        bird.Fly();
        fish.Swim();

        Console.WriteLine("\n=== Проверка команд ===");
        Console.WriteLine($"Собака знает команду 'Сидеть': {dog.KnowsCommand("Сидеть")}");
        Console.WriteLine($"Птица знает команду 'Петь': {bird.KnowsCommand("Петь")}");

        // Демонстрация полиморфизма
        Console.WriteLine("\n=== Полиморфизм через базовый класс ===");
        Animal[] animals = { dog, cat, bird, fish };
        
        foreach (var animal in animals)
        {
            animal.Eat("корм");
            animal.Sleep();
            Console.WriteLine();
        }
````
