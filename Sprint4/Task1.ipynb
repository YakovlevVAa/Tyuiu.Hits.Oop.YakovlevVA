<h1 style="color:DodgerBlue">Явная реализация интерфейса</h1>

В C# интерфейсы могут быть реализованы явным и неявным образом. Давайте подробно рассмотрим, что такое явная реализация интерфейсов, реализация интерфейсов в базовых и производных классах, а также наследование интерфейсов, используя примеры классов `Vehicle`, `Car` и `Motorcycle`.

### Явная реализация интерфейсов

Явная реализация интерфейса используется для того, чтобы скрыть методы интерфейса от стандартного контекста класса. Это означает, что методы интерфейса могут быть вызваны только через ссылку на интерфейс, а не через экземпляр класса.

#### Пример явной реализации

Рассмотрим интерфейс `IVehicle` и его явную реализацию в классе `Car`.

```csharp
public interface IVehicle
{
    void Start();
    void Stop();
    int GetCurrentSpeed();
}

public class Car : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    private int currentSpeed;

    // Явная реализация интерфейса
    void IVehicle.Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    void IVehicle.Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }

    // Неявная реализация метода
    public int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}
```

В этом примере методы `Start` и `Stop` реализованы явно, и для их вызова необходимо использовать объект типа `IVehicle`:

```csharp

IVehicle myCar = new Car { Make = "Toyota", Model = "Camry" };
myCar.Start(); // Корректно, вызывает явную реализацию
// myCar.Stop(); // Корректно, вызывает явную реализацию
```

Если вы попытаетесь вызвать `Start` или `Stop` непосредственно через экземпляр `Car`, это приведёт к ошибке компиляции:

```csharp
Car myCar = new Car();
myCar.Start(); // Ошибка компиляции
```

### Реализация интерфейсов в базовых и производных классах

Интерфейсы могут быть реализованы в базовых классах и наследоваться в производных классах. Давайте рассмотрим эту концепцию на примере, включающем классы `Vehicle`, `Car` и `Motorcycle`.

#### Базовый класс Vehicle

```csharp
public abstract class Vehicle : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    protected int currentSpeed;

    public abstract void Start();  // Оставляем абстрактным
    public abstract void Stop();   // Оставляем абстрактным

    public virtual int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}
```

#### Реализация в производных классах

Теперь реализуем `Car` и `Motorcycle`.

```csharp
public class Car : Vehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }
}

public class Motorcycle : Vehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл остановлен.");
    }
}
```

Теперь оба класса `Car` и `Motorcycle` обязаны реализовать методы `Start` и `Stop`, которые определены в абстрактном классе `Vehicle`.

### Наследование интерфейсов

Интерфейсы могут наследовать друг от друга. Это позволяет создавать сложные иерархии интерфейсов. Например, мы можем создать интерфейс `IMotorizedVehicle`, который наследует `IVehicle`.

```csharp
public interface IMotorizedVehicle : IVehicle
{
    void RevEngine();
}

public class Motorcycle : Vehicle, IMotorizedVehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл остановлен.");
    }

    public void RevEngine()
    {
        Console.WriteLine("Мотоцикл рычит!");
    }
}
```

Таким образом, `Motorcycle` теперь может реализовать все методы из обоих интерфейсов `IVehicle` и `IMotorizedVehicle`.

### Заключение

1. **Явная реализация** интерфейсов скрывает функциональность классов от общего доступа, что помогает избежать путаницы с методами.
   
2. **Реализация интерфейсов в базовых и производных классах** позволяет создавать иерархию классов и обеспечивает обязательность реализации методов в производных классах.

3. **Наследование интерфейсов** позволяет комбинировать несколько интерфейсов, создавая более сложные структуры, что делает код более гибким и расширяемым.

Эти механизмы являются важными инструментами в инструменте разработчика, позволяя создавать более структурированные и управляемые программы.

````




// Базовый интерфейс
public interface IVehicle
{
    void Start();
    void Stop();
    int GetCurrentSpeed();
}

// Интерфейс-наследник добавляет мотора (рев двигателя)
public interface IMotorizedVehicle : IVehicle
{
    void RevEngine();
}




// Производные классы могут переопределять (override) поведение.
public abstract class Vehicle : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    protected int currentSpeed;

    public Vehicle(string make = "Unknown", string model = "Unknown")
    {
        Make = make;
        Model = model;
        currentSpeed = 0;
    }

    // "Hook"-методы, которые производные могут переопределить.
    protected virtual void OnStartCore()
    {
        Console.WriteLine($"{Make} {Model} (Vehicle) базовый запуск.");
    }

    protected virtual void OnStopCore()
    {
        Console.WriteLine($"{Make} {Model} (Vehicle) базовая остановка.");
    }

    // ----- Неявная (implicit) реализация интерфейса: public методы -----
    public virtual void Start()
    {
        currentSpeed = 0;
        OnStartCore();
    }

    public virtual void Stop()
    {
        currentSpeed = 0;
        OnStopCore();
    }

    // Неявная реализация GetCurrentSpeed()
    public virtual int GetCurrentSpeed()
    {
        return currentSpeed;
    }

    public override string ToString() => $"{Make} {Model} (speed={currentSpeed} km/h)";
}


// Car наследует Vehicle и переопределяет поведение (implicit implementation).
public class Car : Vehicle
{
    public int NumberOfDoors { get; set; }

    public Car(string make, string model, int doors = 4)
        : base(make, model)
    {
        NumberOfDoors = doors;
    }

    protected override void OnStartCore()
    {
        Console.WriteLine($"{Make} {Model} (Car) - электрика проверена, двигатель готов.");
    }

    protected override void OnStopCore()
    {
        Console.WriteLine($"{Make} {Model} (Car) - стоянка включена, двигатель выключен.");
    }

    public override void Start()
    {
        base.Start();
        // Дополнительное поведение для Car
        Console.WriteLine($"{Make} {Model} (Car) — плавный старт.");
    }

    public override void Stop()
    {
        base.Stop();
        Console.WriteLine($"{Make} {Model} (Car) — тормоза отпущены.");
    }

    public void Honk() => Console.WriteLine($"{Make} {Model} (Car) — Гудок: Бип-бип!");
}




// Также реализует IMotorizedVehicle (наследует IVehicle).
public class Motorcycle : Vehicle, IMotorizedVehicle
{
    public bool HasSidecar { get; set; }

    public Motorcycle(string make, string model, bool hasSidecar = false)
        : base(make, model)
    {
        HasSidecar = hasSidecar;
    }

    // Публичный override — поведение класса при вызове через тип Motorcycle
    public override void Start()
    {
        // Вызов базовой логики (т.е. "классический" запуск)
        base.Start();
        Console.WriteLine($"{Make} {Model} (Motorcycle) — водитель готов, сцепление включено.");
    }

    public override void Stop()
    {
        base.Stop();
        Console.WriteLine($"{Make} {Model} (Motorcycle) — поставлен на подножку.");
    }

    public override int GetCurrentSpeed()
    {
        // Можно расширить логику
        return base.GetCurrentSpeed();
    }

    public void RevEngine()
    {
        Console.WriteLine($"{Make} {Model} (Motorcycle) — РРРР! (рев двигателя)");
    }



    void IVehicle.Start()
    {
        // Альтернативное поведение при вызове через интерфейс
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} (Motorcycle|IVehicle explicit) — интерфейсный запуск.");
    }

    void IVehicle.Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} (Motorcycle|IVehicle explicit) — интерфейсная остановка.");
    }

    int IVehicle.GetCurrentSpeed()
    {
        // Возвращаем текущее значение
        return currentSpeed;
    }


    void IMotorizedVehicle.RevEngine()
    {
        // Можно переиспользовать публичный метод
        RevEngine();
        Console.WriteLine($"{Make} {Model} (Motorcycle|IMotorizedVehicle explicit) — рев через интерфейс.");
    }
}


Console.WriteLine("=== Демонстрация: неявная реализация (Car) ===");
var car = new Car("Toyota", "Camry", 4);


car.Start();                     // вызывает Car.Start()
Console.WriteLine($"Скорость (class): {car.GetCurrentSpeed()} km/h");

IVehicle carAsVehicle = car;
carAsVehicle.Start();            // также вызывает Car.Start() (неявная реализация)
Console.WriteLine($"Скорость (IVehicle): {carAsVehicle.GetCurrentSpeed()} km/h");

car.Honk();
car.Stop();
Console.WriteLine();

Console.WriteLine("=== Демонстрация: явная реализация (Motorcycle) ===");
var moto = new Motorcycle("Harley-Davidson", "Street");


moto.Start();                    
Console.WriteLine($"Скорость (class): {moto.GetCurrentSpeed()} km/h");


IVehicle motoAsVehicle = moto;
motoAsVehicle.Start();           
Console.WriteLine($"Скорость (IVehicle explicit): {motoAsVehicle.GetCurrentSpeed()} km/h");


IMotorizedVehicle motoMotor = moto;
motoMotor.RevEngine();           


moto.RevEngine();                // публичный метод класса
moto.Stop();
Console.WriteLine();



Console.WriteLine("=== Полиморфизм и коллекция IVehicle / IMotorizedVehicle ===");
var fleet = new List<IVehicle> { car, motoAsVehicle }; 
foreach (var v in fleet)
{
    Console.WriteLine($"Вызов Start() через IVehicle на объекте типа: {v.GetType().Name}");
    v.Start(); 
    Console.WriteLine($"Текущая скорость: {v.GetCurrentSpeed()}");
    v.Stop();
    Console.WriteLine();
}


````
