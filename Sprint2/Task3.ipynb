----

Создать базовый класс Notification в C#, который будет представлять уведомления
пользователям. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
public class Notification
{
    public int NotificationID { get; set; }
    public string MessageText { get; set; }
    public string Type { get; set; }
    // Новые атрибуты
    public DateTime CreatedAt { get; set; }
    public int Priority { get; set; }
    public string Status { get; set; }
    public string Language { get; set; }

    public Notification(int id, string message, string type)
    {
        NotificationID = id;
        MessageText = message;
        Type = type;
        // Инициализация новых атрибутов
        CreatedAt = DateTime.Now;
        Priority = 1;
        Status = "Pending";
        Language = "EN";
    }

    public virtual void SendNotification()
    {
        Console.WriteLine($"Отправка {Type} уведомления...");
        Status = "Sent";
    }

    public virtual void DisplayNotification()
    {
        Console.WriteLine($"Уведомление [{NotificationID}]: {MessageText}");
    }

    public virtual string GetNotificationDetails()
    {
        return $"ID: {NotificationID}, Тип: {Type}, Сообщение: {MessageText}";
    }

    // Новые методы
    public void UpdatePriority(int newPriority)
    {
        Priority = newPriority;
        Console.WriteLine($"Приоритет изменен на {newPriority}");
    }

    public virtual void TranslateNotification(string targetLanguage)
    {
        Console.WriteLine($"Перевод уведомления с {Language} на {targetLanguage}");
        Language = targetLanguage;
    }

    public void MarkAsRead()
    {
        Status = "Read";
        Console.WriteLine("Уведомление помечено как прочитанное");
    }

    public TimeSpan GetAge()
    {
        return DateTime.Now - CreatedAt;
    }
}

// Простое наследование
public class EmailNotification : Notification
{
    public string EmailAddress { get; set; }
    // Новые атрибуты
    public string Subject { get; set; }
    public bool IsHTML { get; set; }
    public string Attachment { get; set; }
    public string CC { get; set; }

    public EmailNotification(int id, string message, string email, string subject = "") 
        : base(id, message, "Email")
    {
        EmailAddress = email;
        Subject = subject;
        IsHTML = true;
    }

    public override void SendNotification()
    {
        base.SendNotification();
        Console.WriteLine($"Отправка email на адрес {EmailAddress} с темой '{Subject}'");
        if (!string.IsNullOrEmpty(Attachment))
            Console.WriteLine($"Прикрепленный файл: {Attachment}");
    }

    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Email: {EmailAddress}, Тема: {Subject}";
    }

    // Новые методы
    public void AddAttachment(string filePath)
    {
        Attachment = filePath;
        Console.WriteLine($"Добавлен вложенный файл: {filePath}");
    }

    public void AddCC(string ccAddress)
    {
        CC = ccAddress;
        Console.WriteLine($"Добавлен CC адрес: {ccAddress}");
    }

    public override void TranslateNotification(string targetLanguage)
    {
        base.TranslateNotification(targetLanguage);
        Console.WriteLine($"Перевод темы письма на {targetLanguage}");
    }
}

// Сложное наследование (многоуровневое)
public class SecureEmailNotification : EmailNotification
{
    public string EncryptionType { get; set; }
    public bool TwoFactorAuth { get; set; }

    public SecureEmailNotification(int id, string message, string email, string encryption) 
        : base(id, message, email, "Secure Notification")
    {
        EncryptionType = encryption;
        TwoFactorAuth = true;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"Применено шифрование: {EncryptionType}");
        if (TwoFactorAuth) Console.WriteLine("Требуется двухфакторная аутентификация");
        base.SendNotification();
    }

    public void EnableTwoFactorAuth()
    {
        TwoFactorAuth = true;
        Console.WriteLine("Двухфакторная аутентификация активирована");
    }
}

// Множественное наследование через интерфейсы
interface IGeolocatable
{
    void SetLocation(double lat, double lon);
    string GetLocation();
}

interface IExpirable
{
    void SetExpiryDate(DateTime expiry);
    bool IsExpired();
}

public class LocationBasedNotification : Notification, IGeolocatable, IExpirable
{
    public double Latitude { get; set; }
    public double Longitude { get; set; }
    public DateTime ExpiryDate { get; set; }
    public string Region { get; set; }

    public LocationBasedNotification(int id, string message, string type) 
        : base(id, message, type) { }

    // Реализация IGeolocatable
    public void SetLocation(double lat, double lon)
    {
        Latitude = lat;
        Longitude = lon;
        Console.WriteLine($"Установлена локация: {lat}, {lon}");
    }

    public string GetLocation()
    {
        return $"Широта: {Latitude}, Долгота: {Longitude}";
    }

    // Реализация IExpirable
    public void SetExpiryDate(DateTime expiry)
    {
        ExpiryDate = expiry;
        Console.WriteLine($"Установлена дата истечения: {expiry}");
    }

    public bool IsExpired()
    {
        return DateTime.Now > ExpiryDate;
    }

    // Новые методы
    public void SetRegion(string region)
    {
        Region = region;
        Console.WriteLine($"Установлен регион: {region}");
    }

    public override void SendNotification()
    {
        base.SendNotification();
        Console.WriteLine($"Локация уведомления: {GetLocation()}");
        if (IsExpired())
            Console.WriteLine("ВНИМАНИЕ: Уведомление просрочено!");
    }
}

// Дополнительный класс для демонстрации
public class SMSNotification : Notification
{
    public string PhoneNumber { get; set; }
    public string SMSSender { get; set; }
    public bool IsUnicode { get; set; }

    public SMSNotification(int id, string message, string number) 
        : base(id, message, "SMS")
    {
        PhoneNumber = number;
        SMSSender = "DefaultSender";
    }

    public override void SendNotification()
    {
        base.SendNotification();
        Console.WriteLine($"Отправка SMS через {SMSSender} на номер {PhoneNumber}");
    }

    public void SetSender(string sender)
    {
        SMSSender = sender;
        Console.WriteLine($"Имя отправителя изменено на: {sender}");
    }
}


Console.WriteLine("Демонстрация системы уведомлений:");
Console.WriteLine("==================================");

var notifications = new List<Notification>
{
    new EmailNotification(1, "Добро пожаловать в систему!", "user@example.com", "Приветствие"),
    new SecureEmailNotification(2, "Конфиденциальные данные", "admin@example.com", "AES-256"),
    new LocationBasedNotification(3, "Спецпредложение в вашем районе", "GeoPush"),
    new SMSNotification(4, "Ваш код подтверждения: 1234", "+79991234567")
};

// Настройка дополнительных параметров
((EmailNotification)notifications[0]).AddAttachment("manual.pdf");
((SecureEmailNotification)notifications[1]).EnableTwoFactorAuth();
((LocationBasedNotification)notifications[2]).SetLocation(55.7558, 37.6173);
((LocationBasedNotification)notifications[2]).SetExpiryDate(DateTime.Now.AddDays(1));
((SMSNotification)notifications[3]).SetSender("MyApp");

foreach (var notification in notifications)
{
    Console.WriteLine("\n--- Обработка уведомления ---");
    notification.DisplayNotification();
    notification.SendNotification();
    notification.UpdatePriority(2);
    notification.TranslateNotification("RU");
    Console.WriteLine("Детали: " + notification.GetNotificationDetails());
    Console.WriteLine($"Возраст уведомления: {notification.GetAge().Minutes} минут");
}

Console.WriteLine("\n==================================");
Console.WriteLine("Обработка всех уведомлений завершена!");
```
