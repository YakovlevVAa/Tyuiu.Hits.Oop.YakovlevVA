<h1 style="color:DodgerBlue">Generic классы в C#</h1>


Генерические классы в C# представляют собой шаблон класса, который может работать с различными типами данных без потери производительности. Они позволяют создавать компоненты, которые могут быть безопасно повторно использованы в различных контекстах, обеспечивая при этом строгую типизацию и избегая необходимости приведения типов.

#### Основные преимущества генерических классов:

1. **Безопасность типов**: Генерические классы обеспечивают проверку типов на этапе компиляции, что помогает предотвратить ошибки во время выполнения.
2. **Универсальность**: Генерические классы могут работать с различными типами данных, делая их очень гибкими и универсальными.
3. **Производительность**: Генерические классы не приводят к значительной потере производительности, поскольку информация о типе хранится вместе с объектом.

#### Пример с классами `Vehicle`, `Car` и `Motorcycle`

Давайте рассмотрим пример генерического класса `VehicleCollection<T>`, который может хранить коллекцию объектов типа `T`, где `T` может быть любой строкой, например, `Vehicle`, `Car` или `Motorcycle`.

```csharp
using System;
using System.Collections.Generic;

// Определение базового класса Vehicle
public class Vehicle
{
    public virtual void Drive()
    {
        Console.WriteLine("Транспортное средство находится в движении.");
    }
}

// Определение производного класса Car
public class Car : Vehicle
{
    public override void Drive()
    {
        Console.WriteLine("Машина едет.");
    }
}

// Определение производного класса Motorcycle
public class Motorcycle : Vehicle
{
    public override void Drive()
    {
        Console.WriteLine("Мотоцикл едет.");
    }
}

// Определение генерического класса VehicleCollection<T>
public class VehicleCollection<T> where T : Vehicle
{
    private List<T> _vehicles = new List<T>();

    public void Add(T vehicle)
    {
        _vehicles.Add(vehicle);
    }

    public void Remove(T vehicle)
    {
        _vehicles.Remove(vehicle);
    }

    public void DisplayVehicles()
    {
        foreach (var vehicle in _vehicles)
        {
            vehicle.Drive(); // Вызов метода Drive() для каждого транспортного средства
        }
    }
}


VehicleCollection<Car> cars = new VehicleCollection<Car>();
cars.Add(new Car());
cars.Add(new Motorcycle()); // Это возможно благодаря ограничению "where T : Vehicle"

cars.DisplayVehicles(); // Выводит: "Машина едет." и "Мотоцикл едет."


```

В этом примере `VehicleCollection<T>` является генерическим классом, который может хранить коллекцию объектов типа `T`, где `T` должен быть подклассом `Vehicle`. Это позволяет создавать коллекции автомобилей, мотоциклов и других видов транспортных средств, обеспечивая при этом строгую типизацию и безопасность типов.

````


// Базовый класс Vehicle
public class Vehicle
{
    public string Brand { get; set; }
    
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Транспортное средство марки {Brand}.");
    }
}

// Производный класс Car
public class Car : Vehicle
{
    public int Doors { get; set; }
    
    public override void DisplayInfo()
    {
        Console.WriteLine($"Автомобиль {Brand} с {Doors} дверями.");
    }
}

// Производный класс Motorcycle
public class Motorcycle : Vehicle
{
    public bool HasSidecar { get; set; }
    
    public override void DisplayInfo()
    {
        Console.WriteLine($"Мотоцикл {Brand}. {(HasSidecar ? "С боковой коляской" : "Без боковой коляски")}");
    }
}

// Генерическая коллекция с ограничением where T : Vehicle
public class VehicleCollection<T> where T : Vehicle
{
    private List<T> _vehicles = new List<T>();
    
    public void AddVehicle(T vehicle)
    {
        _vehicles.Add(vehicle);
    }
    
    public void RemoveVehicle(T vehicle)
    {
        _vehicles.Remove(vehicle);
    }
    
    public void ShowAllVehicles()
    {
        foreach (var vehicle in _vehicles)
        {
            vehicle.DisplayInfo();
        }
    }
}


        // Пример использования с Car
        VehicleCollection<Car> carCollection = new VehicleCollection<Car>();
        
        carCollection.AddVehicle(new Car { Brand = "Toyota", Doors = 4 });
        carCollection.AddVehicle(new Car { Brand = "Honda", Doors = 2 });
        
        Console.WriteLine("Коллекция автомобилей:");
        carCollection.ShowAllVehicles();
        
        // Пример использования с Motorcycle
        VehicleCollection<Motorcycle> motorcycleCollection = new VehicleCollection<Motorcycle>();
        
        motorcycleCollection.AddVehicle(new Motorcycle { Brand = "Harley-Davidson", HasSidecar = true });
        motorcycleCollection.AddVehicle(new Motorcycle { Brand = "Yamaha", HasSidecar = false });
        
        Console.WriteLine("\nКоллекция мотоциклов:");
        motorcycleCollection.ShowAllVehicles();
        
        // Пример с базовым типом Vehicle
        VehicleCollection<Vehicle> universalCollection = new VehicleCollection<Vehicle>();
        
        universalCollection.AddVehicle(new Car { Brand = "Ford", Doors = 4 });
        universalCollection.AddVehicle(new Motorcycle { Brand = "Kawasaki", HasSidecar = false });
        universalCollection.AddVehicle(new Vehicle { Brand = "Generic Vehicle" });
        
        Console.WriteLine("\nУниверсальная коллекция:");
        universalCollection.ShowAllVehicles();
````

````

// Базовый класс Animal
public class Animal
{
    private string _name;
    private int _age;
    private double _weight;

    public string Name
    {
        get { return _name; }
        set 
        { 
            if (!string.IsNullOrWhiteSpace(value))
                _name = value;
            else
                throw new ArgumentException("Имя не может быть пустым");
        }
    }

    public int Age
    {
        get { return _age; }
        set 
        { 
            if (value >= 0)
                _age = value;
            else
                throw new ArgumentException("Возраст не может быть отрицательным");
        }
    }

    public double Weight
    {
        get { return _weight; }
        set 
        { 
            if (value > 0)
                _weight = value;
            else
                throw new ArgumentException("Вес должен быть положительным");
        }
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издает звук.");
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Животное: {Name}, Возраст: {Age} лет, Вес: {Weight} кг");
    }
}

// Производный класс Dog
public class Dog : Animal
{
    public string Breed { get; set; }
    public bool IsTrained { get; set; }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} гавкает: Гав-гав!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}, Дрессирован: {(IsTrained ? "Да" : "Нет")}");
    }

    public void Fetch()
    {
        Console.WriteLine($"{Name} приносит палку!");
    }
}

// Производный класс Cat
public class Cat : Animal
{
    public string Color { get; set; }
    public bool IsIndoor { get; set; }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Цвет: {Color}, Домашний: {(IsIndoor ? "Да" : "Нет")}");
    }

    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: Мрррр...");
    }
}

// Производный класс Bird
public class Bird : Animal
{
    public double Wingspan { get; set; }
    public bool CanFly { get; set; }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {Wingspan} см, Умеет летать: {(CanFly ? "Да" : "Нет")}");
    }

    public void Fly()
    {
        if (CanFly)
            Console.WriteLine($"{Name} летит в небе!");
        else
            Console.WriteLine($"{Name} не умеет летать.");
    }
}

// Производный класс Fish
public class Fish : Animal
{
    public string WaterType { get; set; }
    public bool HasScales { get; set; }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} издает булькающие звуки...");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип воды: {WaterType}, Чешуя: {(HasScales ? "Есть" : "Нет")}");
    }

    public void Swim()
    {
        Console.WriteLine($"{Name} плавает в {WaterType} воде.");
    }
}

// Генерический класс ZooEnclosure для содержания животных
public class ZooEnclosure<T> where T : Animal
{
    private List<T> _animals = new List<T>();
    private string _enclosureName;

    public ZooEnclosure(string name)
    {
        _enclosureName = name;
    }

    public string EnclosureName => _enclosureName;

    public void AddAnimal(T animal)
    {
        _animals.Add(animal);
        Console.WriteLine($"{animal.Name} добавлен(а) в вольер '{_enclosureName}'");
    }

    public bool RemoveAnimal(string animalName)
    {
        var animal = _animals.Find(a => a.Name == animalName);
        if (animal != null)
        {
            _animals.Remove(animal);
            Console.WriteLine($"{animalName} удален(а) из вольера '{_enclosureName}'");
            return true;
        }
        return false;
    }

    public void DisplayAllAnimals()
    {
        Console.WriteLine($"\n=== Вольер '{_enclosureName}' ===");
        if (_animals.Count == 0)
        {
            Console.WriteLine("Вольер пуст");
            return;
        }

        foreach (var animal in _animals)
        {
            animal.DisplayInfo();
            Console.WriteLine("---");
        }
    }

    public void MakeAllSounds()
    {
        Console.WriteLine($"\nЗвуки из вольера '{_enclosureName}':");
        foreach (var animal in _animals)
        {
            animal.MakeSound();
        }
    }

    public T FindAnimal(string name)
    {
        return _animals.Find(a => a.Name == name);
    }

    public int AnimalCount => _animals.Count;
}


        // Создаем специализированные вольеры для разных типов животных
        ZooEnclosure<Dog> dogKennel = new ZooEnclosure<Dog>("Собачий вольер");
        ZooEnclosure<Cat> catHouse = new ZooEnclosure<Cat>("Кошачий дом");
        ZooEnclosure<Bird> aviary = new ZooEnclosure<Bird>("Птичий вольер");
        ZooEnclosure<Fish> aquarium = new ZooEnclosure<Fish>("Аквариум");

        // Универсальный вольер для любых животных
        ZooEnclosure<Animal> mixedEnclosure = new ZooEnclosure<Animal>("Смешанный вольер");

        // Создаем животных
        Dog dog1 = new Dog { Name = "Бобик", Age = 3, Weight = 12.5, Breed = "Овчарка", IsTrained = true };
        Dog dog2 = new Dog { Name = "Шарик", Age = 5, Weight = 8.2, Breed = "Дворняжка", IsTrained = false };

        Cat cat1 = new Cat { Name = "Мурка", Age = 2, Weight = 4.1, Color = "Рыжий", IsIndoor = true };
        Cat cat2 = new Cat { Name = "Барсик", Age = 4, Weight = 5.3, Color = "Серый", IsIndoor = false };

        Bird bird1 = new Bird { Name = "Кеша", Age = 1, Weight = 0.3, Wingspan = 25.5, CanFly = true };
        Bird bird2 = new Bird { Name = "Гоша", Age = 2, Weight = 1.2, Wingspan = 15.0, CanFly = false };

        Fish fish1 = new Fish { Name = "Немо", Age = 1, Weight = 0.1, WaterType = "морская", HasScales = true };
        Fish fish2 = new Fish { Name = "Дори", Age = 2, Weight = 0.2, WaterType = "морская", HasScales = true };

        // Добавляем животных в соответствующие вольеры
        dogKennel.AddAnimal(dog1);
        dogKennel.AddAnimal(dog2);

        catHouse.AddAnimal(cat1);
        catHouse.AddAnimal(cat2);

        aviary.AddAnimal(bird1);
        aviary.AddAnimal(bird2);

        aquarium.AddAnimal(fish1);
        aquarium.AddAnimal(fish2);

        // Добавляем разных животных в смешанный вольер
        mixedEnclosure.AddAnimal(new Dog { Name = "Рекс", Age = 4, Weight = 10.0, Breed = "Лабрадор", IsTrained = true });
        mixedEnclosure.AddAnimal(new Cat { Name = "Васька", Age = 3, Weight = 4.5, Color = "Черный", IsIndoor = true });
        mixedEnclosure.AddAnimal(new Bird { Name = "Соловей", Age = 1, Weight = 0.2, Wingspan = 20.0, CanFly = true });

        // Демонстрация работы с вольерами
        Console.WriteLine("=== ИНФОРМАЦИЯ О ВОЛЬЕРАХ ===");
        
        dogKennel.DisplayAllAnimals();
        dogKennel.MakeAllSounds();

        catHouse.DisplayAllAnimals();
        catHouse.MakeAllSounds();

        aviary.DisplayAllAnimals();
        aviary.MakeAllSounds();

        aquarium.DisplayAllAnimals();
        aquarium.MakeAllSounds();

        mixedEnclosure.DisplayAllAnimals();
        mixedEnclosure.MakeAllSounds();

        // Демонстрация специальных методов
        Console.WriteLine("\n=== СПЕЦИАЛЬНЫЕ ВОЗМОЖНОСТИ ===");
        
        var foundDog = dogKennel.FindAnimal("Бобик");
        if (foundDog != null)
            foundDog.Fetch();

        var foundCat = catHouse.FindAnimal("Мурка");
        if (foundCat != null)
            ((Cat)foundCat).Purr();

        var foundBird = aviary.FindAnimal("Кеша");
        if (foundBird != null)
            ((Bird)foundBird).Fly();

        var foundFish = aquarium.FindAnimal("Немо");
        if (foundFish != null)
            ((Fish)foundFish).Swim();

        // Статистика
        Console.WriteLine($"\n=== СТАТИСТИКА ЗООПАРКА ===");
        Console.WriteLine($"Собак: {dogKennel.AnimalCount}");
        Console.WriteLine($"Кошек: {catHouse.AnimalCount}");
        Console.WriteLine($"Птиц: {aviary.AnimalCount}");
        Console.WriteLine($"Рыб: {aquarium.AnimalCount}");
        Console.WriteLine($"Всего животных: {dogKennel.AnimalCount + catHouse.AnimalCount + aviary.AnimalCount + aquarium.AnimalCount + mixedEnclosure.AnimalCount}");
````
