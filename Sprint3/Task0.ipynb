### Понятие полиморфизма в C# с параметрами

Полиморфизм в объектно-ориентированном программировании позволяет объектам разных классов обрабатывать одни и те же сообщения по-разному. В C#, полиморфизм часто реализуется через наследование и переопределение методов. Давайте рассмотрим пример с классами `Vehicle`, `Car` и `Motorcycle`, чтобы продемонстрировать, как полиморфизм работает с параметрами.

#### Определение классов

Сначала определим базовый класс `Vehicle` с виртуальным методом `Drive()`, который принимает параметры. Затем создадим два производных класса: `Car` и `Motorcycle`, которые переопределят этот метод.

```csharp
// Базовый класс Vehicle с виртуальным методом Drive(), принимающим параметры
public class Vehicle
{
    public virtual void Drive(int speed)
    {
        Console.WriteLine($"Транспорт движется {speed} km/h.");
    }
}

// Производный класс Car, переопределяющий метод Drive()
public class Car : Vehicle
{
    public override void Drive(int speed)
    {
        Console.WriteLine($"Автомобиль движется {speed} km/h.");
    }
}

// Производный класс Motorcycle, также переопределяющий метод Drive()
public class Motorcycle : Vehicle
{
    public override void Drive(int speed)
    {
        Console.WriteLine($"Мотоцикл движется {speed} km/h.");
    }
}
```

#### Использование полиморфизма

Теперь создадим список объектов типа `Vehicle` и добавим в него экземпляры `Car` и `Motorcycle`. Затем вызовем метод `Drive()` для каждого элемента списка, передавая разные скорости. Это демонстрирует, как полиморфизм позволяет нам работать с объектами разных типов через общий интерфейс, но при этом выполнять различные действия в зависимости от "реального" типа объекта.

```csharp
List<Vehicle> vehicles = new List<Vehicle>
{
    new Car(),
    new Motorcycle(),
    new Vehicle() // Добавляем экземпляр базового класса для демонстрации
};

foreach (var vehicle in vehicles)
{
    vehicle.Drive(100); // Вызываем метод Drive() 
}

```

В результате выполнения этого кода, каждый объект в списке вызывает свою версию метода `Drive()`, демонстрируя, что при работе с объектами через базовый класс (`Vehicle`) можно достичь различных действий в зависимости от "реального" типа объекта, даже когда используются разные скорости.

Этот пример иллюстрирует ключевое понятие полиморфизма в ООП: возможность работы с объектами разных типов через общий интерфейс или базовый класс, что позволяет писать более гибкий и расширяемый код.

````
// Базовый класс Vehicle
public class Vehicle
{
    public virtual void Drive(int speed, string roadType)
    {
        Console.WriteLine($"Транспорт движется {speed} км/ч по {roadType}.");
    }
}

// Производный класс Car
public class Car : Vehicle
{
    public override void Drive(int speed, string roadType)
    {
        if (roadType == "бездорожью")
            Console.WriteLine($"Автомобиль с трудом преодолевает бездорожье на {speed} км/ч");
        else
            Console.WriteLine($"Автомобиль плавно едет {speed} км/ч по {roadType}");
    }
}

// Производный класс Motorcycle
public class Motorcycle : Vehicle
{
    public override void Drive(int speed, string roadType)
    {
        if (roadType == "гравию")
            Console.WriteLine($"Мотоцикл осторожно маневрирует на {speed} км/ч по гравию");
        else
            Console.WriteLine($"Мотоцикл быстро мчится {speed} км/ч по {roadType}");
    }
}

        // Создаем список разных транспортных средств
        List<Vehicle> vehicles = new List<Vehicle>
        {
            new Car(),
            new Motorcycle(),
            new Vehicle()
        };

        // Тестируем разные сценарии с параметрами
        int[] speeds = { 60, 80, 40 };
        string[] roadTypes = { "асфальту", "гравию", "бездорожью" };

        // Демонстрируем полиморфное поведение
        for (int i = 0; i < vehicles.Count; i++)
        {
            vehicles[i].Drive(speeds[i], roadTypes[i]);
            Console.WriteLine(new string('-', 50));
        }

        // Дополнительный пример с одинаковыми параметрами
        Console.WriteLine("Одинаковые параметры для всех:");
        foreach (var vehicle in vehicles)
        {
            vehicle.Drive(70, "асфальту");
        }
````

````

// Базовый класс Animal
public class Animal
{
    // Поля с защищенным доступом
    protected string _name;
    protected int _age;
    protected double _weight;
    
    // Конструктор базового класса
    public Animal(string name, int age, double weight)
    {
        _name = name;
        _age = age;
        _weight = weight;
    }
    
    // Свойства с геттерами и сеттерами
    public string Name
    {
        get { return _name; }
        set 
        { 
            if (!string.IsNullOrWhiteSpace(value))
                _name = value;
            else
                throw new ArgumentException("Имя не может быть пустым");
        }
    }
    
    public int Age
    {
        get { return _age; }
        set 
        { 
            if (value >= 0)
                _age = value;
            else
                throw new ArgumentException("Возраст не может быть отрицательным");
        }
    }
    
    public double Weight
    {
        get { return _weight; }
        set 
        { 
            if (value > 0)
                _weight = value;
            else
                throw new ArgumentException("Вес должен быть положительным");
        }
    }
    
    // Виртуальный метод для демонстрации полиморфизма
    public virtual void MakeSound()
    {
        Console.WriteLine($"{_name} издает звук");
    }
    
    // Виртуальный метод с параметрами
    public virtual void Eat(string food, int quantity)
    {
        Console.WriteLine($"{_name} съедает {quantity} единиц {food}");
        _weight += quantity * 0.1; // Увеличиваем вес в зависимости от количества еды
    }
    
    // Виртуальный метод для получения информации
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Животное: {_name}, Возраст: {_age} лет, Вес: {_weight} кг");
    }
}

// Производный класс Dog
public class Dog : Animal
{
    // Дополнительное свойство только для собак
    public string Breed { get; set; }
    
    // Конструктор
    public Dog(string name, int age, double weight, string breed) : base(name, age, weight)
    {
        Breed = breed;
    }
    
    // Переопределение метода MakeSound
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} породы {Breed} громко лает: Гав-гав!");
    }
    
    // Переопределение метода Eat с дополнительной логикой
    public override void Eat(string food, int quantity)
    {
        if (food.ToLower() == "кость")
        {
            Console.WriteLine($"{Name} с радостью грызет {quantity} костей!");
            Weight += quantity * 0.2;
        }
        else
        {
            Console.WriteLine($"{Name} неохотно ест {quantity} единиц {food}");
            Weight += quantity * 0.1;
        }
    }
    
    // Переопределение метода DisplayInfo
    public override void DisplayInfo()
    {
        Console.WriteLine($"Собака: {Name}, Порода: {Breed}, Возраст: {Age} лет, Вес: {Weight} кг");
    }
    
    // Уникальный метод для класса Dog
    public void Fetch(string item)
    {
        Console.WriteLine($"{Name} бежит за {item} и приносит его обратно!");
    }
}

// Производный класс Cat
public class Cat : Animal
{
    // Дополнительное свойство
    public bool IsIndoor { get; set; }
    
    // Конструктор
    public Cat(string name, int age, double weight, bool isIndoor) : base(name, age, weight)
    {
        IsIndoor = isIndoor;
    }
    
    // Переопределение метода MakeSound
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мурлычет: Мяу-мяу! {(IsIndoor ? "(домашний кот)" : "(уличный кот)")}");
    }
    
    // Переопределение метода Eat
    public override void Eat(string food, int quantity)
    {
        if (food.ToLower() == "рыба")
        {
            Console.WriteLine($"{Name} с удовольствием съедает {quantity} рыб!");
            Weight += quantity * 0.15;
        }
        else if (food.ToLower() == "молоко")
        {
            Console.WriteLine($"{Name} лакает {quantity} мисок молока");
            Weight += quantity * 0.05;
        }
        else
        {
            base.Eat(food, quantity);
        }
    }
    
    // Переопределение метода DisplayInfo
    public override void DisplayInfo()
    {
        string type = IsIndoor ? "Домашний" : "Уличный";
        Console.WriteLine($"Кот: {Name}, {type}, Возраст: {Age} лет, Вес: {Weight} кг");
    }
    
    // Уникальный метод для класса Cat
    public void Climb(string objectName)
    {
        Console.WriteLine($"{Name} ловко забирается на {objectName}");
    }
}

// Производный класс Bird
public class Bird : Animal
{
    // Дополнительные свойства
    public double Wingspan { get; set; }
    public bool CanFly { get; set; }
    
    // Конструктор
    public Bird(string name, int age, double weight, double wingspan, bool canFly) : base(name, age, weight)
    {
        Wingspan = wingspan;
        CanFly = canFly;
    }
    
    // Переопределение метода MakeSound
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик! Размах крыльев: {Wingspan} см");
    }
    
    // Переопределение метода Eat
    public override void Eat(string food, int quantity)
    {
        if (food.ToLower() == "зерно")
        {
            Console.WriteLine($"{Name} клюет {quantity} зернышек");
            Weight += quantity * 0.02;
        }
        else if (food.ToLower() == "насекомые")
        {
            Console.WriteLine($"{Name} ловит {quantity} насекомых");
            Weight += quantity * 0.01;
        }
        else
        {
            Console.WriteLine($"{Name} осторожно пробует {food}");
            Weight += quantity * 0.01;
        }
    }
    
    // Переопределение метода DisplayInfo
    public override void DisplayInfo()
    {
        string flightAbility = CanFly ? "Летающая" : "Нелетающая";
        Console.WriteLine($"Птица: {Name}, {flightAbility}, Размах крыльев: {Wingspan} см, Возраст: {Age} лет, Вес: {Weight} кг");
    }
    
    // Уникальный метод для класса Bird
    public void Fly(int distance)
    {
        if (CanFly)
        {
            Console.WriteLine($"{Name} пролетает {distance} метров в небе!");
        }
        else
        {
            Console.WriteLine($"{Name} пытается взлететь, но не может летать");
        }
    }
}

// Производный класс Fish
public class Fish : Animal
{
    // Дополнительные свойства
    public string WaterType { get; set; } // "пресная" или "соленая"
    public bool HasScales { get; set; }
    
    // Конструктор
    public Fish(string name, int age, double weight, string waterType, bool hasScales) : base(name, age, weight)
    {
        WaterType = waterType;
        HasScales = hasScales;
    }
    
    // Переопределение метода MakeSound
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} пускает пузыри: Буль-буль! ({WaterType} вода)");
    }
    
    // Переопределение метода Eat
    public override void Eat(string food, int quantity)
    {
        if (food.ToLower() == "планктон")
        {
            Console.WriteLine($"{Name} фильтрует {quantity} планктона из воды");
            Weight += quantity * 0.001;
        }
        else if (food.ToLower() == "водоросли")
        {
            Console.WriteLine($"{Name} nibbles on {quantity} водорослей");
            Weight += quantity * 0.005;
        }
        else
        {
            Console.WriteLine($"{Name} игнорирует {food} и ищет другую еду");
        }
    }
    
    // Переопределение метода DisplayInfo
    public override void DisplayInfo()
    {
        string scales = HasScales ? "Чешуйчатая" : "Без чешуи";
        Console.WriteLine($"Рыба: {Name}, {scales}, {WaterType} вода, Возраст: {Age} лет, Вес: {Weight} кг");
    }
    
    // Уникальный метод для класса Fish
    public void Swim(int depth)
    {
        Console.WriteLine($"{Name} плавает на глубине {depth} метров");
    }
}

        // Создаем список животных разных типов
        List<Animal> animals = new List<Animal>
        {
            new Dog("Бобик", 3, 12.5, "Овчарка"),
            new Cat("Мурзик", 2, 4.2, true),
            new Bird("Кеша", 1, 0.3, 25.5, true),
            new Fish("Немо", 1, 0.5, "соленая", true),
            new Dog("Шарик", 5, 8.7, "Дворняжка"),
            new Bird("Страус", 4, 45.0, 200.0, false)
        };
        
        Console.WriteLine("=== ПОЛИМОРФИЗМ ===");
        Console.WriteLine();
        
        // Демонстрация полиморфного поведения
        Console.WriteLine("1. Все животные издают звуки:");
        Console.WriteLine(new string('-', 40));
        foreach (var animal in animals)
        {
            animal.MakeSound(); // Полиморфный вызов
        }
        
        Console.WriteLine();
        Console.WriteLine("2. Кормление животных (разная реакция на разную еду):");
        Console.WriteLine(new string('-', 40));
        
        // Кормим животных разной едой
        animals[0].Eat("кость", 2);        // Собака
        animals[1].Eat("рыба", 1);         // Кот
        animals[2].Eat("зерно", 10);       // Птица
        animals[3].Eat("планктон", 100);   // Рыба
        animals[4].Eat("овощи", 3);        // Другая собака
        animals[5].Eat("насекомые", 5);    // Страус
        
        Console.WriteLine();
        Console.WriteLine("3. Информация о всех животных:");
        Console.WriteLine(new string('-', 40));
        foreach (var animal in animals)
        {
            animal.DisplayInfo(); // Полиморфный вызов
        }
        
        Console.WriteLine();
        Console.WriteLine("4. Уникальные способности каждого типа животных:");
        Console.WriteLine(new string('-', 40));
        
        // Демонстрация уникальных методов через приведение типов
        foreach (var animal in animals)
        {
            if (animal is Dog dog)
            {
                dog.Fetch("палку");
            }
            else if (animal is Cat cat)
            {
                cat.Climb("дерево");
            }
            else if (animal is Bird bird)
            {
                bird.Fly(50);
            }
            else if (animal is Fish fish)
            {
                fish.Swim(10);
            }
        }
        
        Console.WriteLine();
        Console.WriteLine("5. Работа со свойствами через геттеры и сеттеры:");
        Console.WriteLine(new string('-', 40));
        
        // Изменяем свойства и показ валидации
        try
        {
            animals[0].Name = "Рекс"; // Корректное имя
            animals[0].Age = 4;       // Корректный возраст
            Console.WriteLine($"Собака теперь зовут: {animals[0].Name}, возраст: {animals[0].Age}");
            
        }
        catch (ArgumentException ex)
        {
            Console.WriteLine($"Ошибка: {ex.Message}");
        }
````
