Создать базовый класс Notification в C#, который будет представлять уведомления
пользователям. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.
### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

```
public class Notification
{
    public int NotificationID { get; set; }
    public string MessageText { get; set; }
    public string Type { get; set; }
    public DateTime CreatedDate { get; set; }
    public int Priority { get; set; }
    public bool IsRead { get; set; }
    public string Language { get; set; }

    public Notification(int id, string message, string type)
    {
        NotificationID = id;
        MessageText = message;
        Type = type;
        CreatedDate = DateTime.Now;
        Priority = 1;
        IsRead = false;
        Language = "RU";
    }

    // Перегрузка конструктора
    public Notification(int id, string message, string type, int priority, string language)
        : this(id, message, type)
    {
        Priority = priority;
        Language = language;
    }

    public virtual void SendNotification()
    {
        Console.WriteLine($"Отправка {Type} уведомления...");
        MarkAsSent();
    }

    public virtual void DisplayNotification()
    {
        Console.WriteLine($"Уведомление [{NotificationID}]: {MessageText}");
    }

    // Перегруженный метод
    public void DisplayNotification(bool showDetails)
    {
        if (showDetails)
        {
            Console.WriteLine($"Уведомление [{NotificationID}]: {MessageText}");
            Console.WriteLine($"Приоритет: {Priority}, Язык: {Language}");
        }
        else
        {
            DisplayNotification();
        }
    }

    public virtual string GetNotificationDetails()
    {
        return $"ID: {NotificationID}, Тип: {Type}, Сообщение: {MessageText}, Приоритет: {Priority}";
    }

    // Новые методы
    public void MarkAsRead()
    {
        IsRead = true;
        Console.WriteLine($"Уведомление {NotificationID} помечено как прочитанное");
    }

    public void UpdatePriority(int newPriority)
    {
        Priority = newPriority;
        Console.WriteLine($"Приоритет уведомления {NotificationID} изменен на {newPriority}");
    }

    protected virtual void MarkAsSent()
    {
        Console.WriteLine($"Уведомление {NotificationID} помечено как отправленное");
    }

    // Перегруженный метод
    public virtual void SendNotification(string recipient)
    {
        Console.WriteLine($"Отправка {Type} уведомления для {recipient}...");
    }
}

public class EmailNotification : Notification
{
    public string EmailAddress { get; set; }
    public string Subject { get; set; }
    public string Attachment { get; set; }
    public bool IsHTML { get; set; }

    public EmailNotification(int id, string message, string email) 
        : base(id, message, "Email")
    {
        EmailAddress = email;
        Subject = "Уведомление";
        IsHTML = false;
    }

    // Перегрузка конструктора
    public EmailNotification(int id, string message, string email, string subject, bool isHtml)
        : base(id, message, "Email")
    {
        EmailAddress = email;
        Subject = subject;
        IsHTML = isHtml;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"Отправка email на адрес {EmailAddress}");
        Console.WriteLine($"Тема: {Subject}, HTML: {IsHTML}");
        Console.WriteLine($"Текст: {MessageText}");
        if (!string.IsNullOrEmpty(Attachment))
            Console.WriteLine($"Вложение: {Attachment}");
        MarkAsSent();
    }

    // Переопределение с новой функциональностью
    public override void SendNotification(string recipient)
    {
        EmailAddress = recipient;
        Console.WriteLine($"Отправка email на адрес {recipient} с темой '{Subject}'");
    }

    public void AddAttachment(string filePath)
    {
        Attachment = filePath;
        Console.WriteLine($"Добавлено вложение: {filePath}");
    }

    public void ConvertToHTML()
    {
        IsHTML = true;
        Console.WriteLine("Email преобразован в HTML формат");
    }

    protected override void MarkAsSent()
    {
        Console.WriteLine($"Email уведомление {NotificationID} доставлено на {EmailAddress}");
    }
}

public class SMSNotification : Notification
{
    public string PhoneNumber { get; set; }
    public string SenderName { get; set; }
    public bool IsUnicode { get; set; }
    public int MessageCount { get; set; }

    public SMSNotification(int id, string message, string number) 
        : base(id, message, "SMS")
    {
        PhoneNumber = number;
        SenderName = "System";
        IsUnicode = false;
        MessageCount = 1;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"Отправка SMS на номер {PhoneNumber} через отправителя '{SenderName}'");
        Console.WriteLine($"Текст: {MessageText}");
        Console.WriteLine($"Количество сообщений: {MessageCount}, Unicode: {IsUnicode}");
        MarkAsSent();
    }

    public void SetSender(string sender)
    {
        SenderName = sender;
        Console.WriteLine($"Имя отправителя установлено: {sender}");
    }

    public void CalculateMessageCount()
    {
        MessageCount = (int)Math.Ceiling(MessageText.Length / 160.0);
        Console.WriteLine($"Рассчитано количество SMS: {MessageCount}");
    }

    protected override void MarkAsSent()
    {
        Console.WriteLine($"SMS уведомление {NotificationID} отправлено на {PhoneNumber}");
    }
}

public class PushNotification : Notification
{
    public string Platform { get; set; }
    public string DeviceToken { get; set; }
    public int ExpiryHours { get; set; }
    public string Sound { get; set; }

    public PushNotification(int id, string message, string platform) 
        : base(id, message, "Push")
    {
        Platform = platform;
        DeviceToken = "default_token";
        ExpiryHours = 24;
        Sound = "default";
    }

    public override void DisplayNotification()
    {
        Console.WriteLine($"[{Platform}] Push-уведомление: {MessageText}");
        Console.WriteLine($"Звук: {Sound}, Срок действия: {ExpiryHours} часов");
    }

    public override void SendNotification()
    {
        Console.WriteLine($"Отправка push-уведомления на платформу {Platform}");
        Console.WriteLine($"Токен устройства: {DeviceToken}");
        Console.WriteLine($"Текст: {MessageText}");
        MarkAsSent();
    }

    public void SetCustomSound(string soundName)
    {
        Sound = soundName;
        Console.WriteLine($"Установлен пользовательский звук: {soundName}");
    }

    public void ExtendExpiry(int hours)
    {
        ExpiryHours += hours;
        Console.WriteLine($"Срок действия увеличен до {ExpiryHours} часов");
    }

    protected override void MarkAsSent()
    {
        Console.WriteLine($"Push уведомление {NotificationID} доставлено на {Platform}");
    }
}

// Generic-класс для работы с уведомлениями
public class NotificationManager<T> where T : Notification
{
    private List<T> _notifications = new List<T>();
    
    public void AddNotification(T notification)
    {
        _notifications.Add(notification);
        Console.WriteLine($"Добавлено уведомление типа {typeof(T).Name} с ID {notification.NotificationID}");
    }
    
    public void SendAllNotifications()
    {
        Console.WriteLine($"\nОтправка всех уведомлений ({_notifications.Count} шт.):");
        foreach (var notification in _notifications)
        {
            notification.SendNotification();
        }
    }
    
    public void DisplayAllNotifications()
    {
        Console.WriteLine($"\nВсе уведомления ({_notifications.Count} шт.):");
        foreach (var notification in _notifications)
        {
            notification.DisplayNotification();
        }
    }
    
    public T FindNotificationById(int id)
    {
        return _notifications.FirstOrDefault(n => n.NotificationID == id);
    }
    
    public List<T> GetHighPriorityNotifications(int minPriority = 2)
    {
        return _notifications.Where(n => n.Priority >= minPriority).ToList();
    }
    
    public void MarkAllAsRead()
    {
        foreach (var notification in _notifications)
        {
            notification.MarkAsRead();
        }
    }
    
    public int Count => _notifications.Count;
}

// Демонстрация работы
Console.WriteLine("Демонстрация расширенной системы уведомлений:");
Console.WriteLine("=============================================");

// Создание менеджеров для разных типов уведомлений
var emailManager = new NotificationManager<EmailNotification>();
var smsManager = new NotificationManager<SMSNotification>();
var pushManager = new NotificationManager<PushNotification>();

// Создание и добавление уведомлений
var email1 = new EmailNotification(1, "Добро пожаловать в нашу систему!", "user@example.com");
var email2 = new EmailNotification(2, "Ваш заказ готов к отправке", "client@domain.com", "Информация о заказе", true);
email2.AddAttachment("invoice.pdf");
emailManager.AddNotification(email1);
emailManager.AddNotification(email2);

var sms1 = new SMSNotification(3, "Ваш код подтверждения: 1234", "+79991234567");
sms1.CalculateMessageCount();
sms1.SetSender("MyCompany");
var sms2 = new SMSNotification(4, "Специальное предложение для вас! Посетите наш сайт чтобы узнать подробности", "+79998887766");
sms2.CalculateMessageCount();
smsManager.AddNotification(sms1);
smsManager.AddNotification(sms2);

var push1 = new PushNotification(5, "У вас новое сообщение", "Android");
push1.SetCustomSound("message.mp3");
var push2 = new PushNotification(6, "Обновление приложения доступно", "iOS");
push2.ExtendExpiry(48);
pushManager.AddNotification(push1);
pushManager.AddNotification(push2);

// Демонстрация работы методов
Console.WriteLine("\n--- Работа с Email уведомлениями ---");
emailManager.SendAllNotifications();

Console.WriteLine("\n--- Работа с SMS уведомлениями ---");
smsManager.DisplayAllNotifications();
smsManager.SendAllNotifications();

Console.WriteLine("\n--- Работа с Push уведомлениями ---");
pushManager.SendAllNotifications();

// Демонстрация перегруженных методов
Console.WriteLine("\n--- Демонстрация перегруженных методов ---");
email1.DisplayNotification(true);
email2.SendNotification("newuser@example.com");

// Поиск и работа с конкретным уведомлением
Console.WriteLine("\n--- Поиск уведомлений ---");
var foundSms = smsManager.FindNotificationById(3);
if (foundSms != null)
{
    foundSms.UpdatePriority(3);
    foundSms.MarkAsRead();
}

Console.WriteLine($"\nВсего email уведомлений: {emailManager.Count}");
Console.WriteLine($"Всего SMS уведомлений: {smsManager.Count}");
Console.WriteLine($"Всего push уведомлений: {pushManager.Count}");

Console.WriteLine("\n=============================================");
Console.WriteLine("Обработка всех уведомлений завершена!");
```
