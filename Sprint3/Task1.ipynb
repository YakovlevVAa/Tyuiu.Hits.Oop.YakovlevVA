### Перекрытие методов в C#

Перекрытие методов (или переопределение) — это процесс, при котором производный класс предоставляет новую реализацию метода, уже существующего в его базовом классе. Это позволяет изменить или расширить поведение метода в производном классе без изменения исходного метода в базовом классе. Перекрытие методов является ключевой особенностью полиморфизма в объектно-ориентированном программировании.

#### Пример с классами `Vehicle`, `Car` и `Motorcycle`

Давайте рассмотрим пример с классами `Vehicle`, `Car` и `Motorcycle`, где каждый из этих классов имеет метод `Drive()`, который принимает параметр `speed`.

```csharp
// Базовый класс Vehicle с виртуальным методом Drive()
public class Vehicle
{
    public virtual void Drive(int speed)
    {
        Console.WriteLine($"Транспорт движется {speed} km/h.");
    }
}

// Производный класс Car, переопределяющий метод Drive()
public class Car : Vehicle
{
    public override void Drive(int speed)
    {
        Console.WriteLine($"Автомобиль движется {speed} km/h.");
    }
}

// Производный класс Motorcycle, также переопределяющий метод Drive()
public class Motorcycle : Vehicle
{
    public override void Drive(int speed)
    {
        Console.WriteLine($"Мотоцикл движется {speed} km/h.");
    }
}
```

В этом примере, метод `Drive()` в классе `Car` и `Motorcycle` переопределяет метод `Drive()` из базового класса `Vehicle`. Это означает, что при вызове метода `Drive()` на объекте типа `Car` или `Motorcycle`, будет вызвана соответствующая реализация метода в этих классах, а не в классе `Vehicle`.

#### Отличие перекрытия от перегрузки методов

Перегрузка методов — это процесс определения нескольких методов с одним и тем же именем, но с разным количеством или типом параметров. Например:

```csharp
public class ExampleClass
{
    public void Print(int number)
    {
        Console.WriteLine(number);
    }

    public void Print(double number)
    {
        Console.WriteLine(number.ToString());
    }
}
```

В этом случае, у нас есть два метода с именем `Print`, но с разным типом параметра. Это отличается от перекрытия методов, где метод в производном классе полностью заменяет метод в базовом классе.

#### Вызов методов родителя

Иногда в производном классе может потребоваться вызвать метод родителя. Для этого используется ключевое слово `base`. Например, если мы хотим добавить дополнительную логику в метод `Drive()` класса `Car`, сохраняя при этом оригинальное поведение метода `Drive()` из класса `Vehicle`.

```csharp
public class Car : Vehicle
{
    public override void Drive(int speed)
    {
        base.Drive(speed); // Вызов метода Drive() из класса Vehicle
        Console.WriteLine("Добавляем специфическое поведение автомобиля");
    }
}
```

В этом примере, после вызова `base.Drive(speed)`, выполняется дополнительная логика, специфичная для класса `Car`.


### Перегрузка методов в C#

Перегрузка методов — это способность класса определить несколько методов с одним и тем же именем, но с разным количеством или типом параметров. Это позволяет разработчикам предоставлять различные варианты выполнения одного и того же действия в зависимости от входных данных. Перегрузка методов является одной из форм выражения полиморфизма в объектно-ориентированном программировании.

#### Пример с классами `Vehicle`, `Car` и `Motorcycle`

Давайте рассмотрим пример, где класс `Vehicle` содержит перегруженные методы `Drive()`, которые принимают разные типы параметров.

```csharp
public class Vehicle
{
    // Метод Drive(), принимающий int
    public void Drive(int speed)
    {
        Console.WriteLine($"Транспорт движется {speed} km/h.");
    }

    // Метод Drive(), принимающий string
    public void Drive(string mode)
    {
        Console.WriteLine($"Автомобиль движется в {mode} режиме.");
    }

    // Метод Drive(), принимающий bool
    public void Drive(bool isManual)
    {
        if (isManual)
        {
            Console.WriteLine("Автомобиль переключает передачи вручную.");
        }
        else
        {
            Console.WriteLine("Автомобиль оснащен автоматической коробкой передач.");
        }
    }
}
```

В этом примере класс `Vehicle` определяет три метода с именем `Drive()`, каждый из которых принимает разные типы параметров: `int`, `string` и `bool`. Это позволяет вызывать разные версии метода `Drive()` в зависимости от контекста, например:

```csharp
Vehicle myVehicle = new Vehicle();

myVehicle.Drive(60); // Вызов метода Drive() с int параметром
myVehicle.Drive("Eco"); // Вызов метода Drive() с string параметром
myVehicle.Drive(true); // Вызов метода Drive() с bool параметром
```

#### Отличие перегрузки от переопределения

Отличие перегрузки методов от переопределения заключается в том, что перегрузка методов включает определение нескольких методов с одним именем, но с разным набором параметров, тогда как переопределение методов (или перекрытие) означает предоставление новой реализации метода в производном классе, который уже существует в базовом классе.

#### Вызов методов

При вызове перегруженного метода компилятор выбирает подходящую версию метода на основе типов и количества аргументов, переданных в вызове. Если типы и количество аргументов совпадают с несколькими версиями метода, компилятор выберет ту версию, которая была объявлена первой.

### Заключение

Перегрузка методов в C# позволяет классам предоставлять различные варианты выполнения одного и того же действия, адаптируясь к различным условиям использования. Это мощный инструмент для создания гибких и удобных API, позволяющих клиентам класса использовать его функциональность наиболее эффективно.

Перекрытие методов в C# позволяет производным классам изменять или расширять поведение методов базовых классов, что является ключевой особенностью полиморфизма. Это отличается от перегрузки методов, которая предполагает определение нескольких методов с одинаковым именем, но с разным набором параметров.

````


// Базовый класс Vehicle с перегрузкой и переопределением методов
public class Vehicle
{
    protected string _brand;
    protected int _year;

    // Конструктор
    public Vehicle(string brand, int year)
    {
        _brand = brand;
        _year = year;
    }

    // Свойства с геттерами и сеттерами
    public string Brand
    {
        get { return _brand; }
        set { _brand = value; }
    }

    public int Year
    {
        get { return _year; }
        set 
        { 
            if (value > 1885 && value <= DateTime.Now.Year)
                _year = value;
            else
                throw new ArgumentException("Некорректный год выпуска");
        }
    }

    // === ПЕРЕГРУЗКА МЕТОДОВ ===
    
    // Метод Drive с одним параметром (скорость)
    public virtual void Drive(int speed)
    {
        Console.WriteLine($"{_brand} движется со скоростью {speed} км/ч");
    }

    // Перегруженный метод Drive с двумя параметрами
    public virtual void Drive(int speed, string direction)
    {
        Console.WriteLine($"{_brand} движется {direction} со скоростью {speed} км/ч");
    }

    // Перегруженный метод Drive с тремя параметрами
    public virtual void Drive(int speed, string direction, int time)
    {
        Console.WriteLine($"{_brand} движется {direction} со скоростью {speed} км/ч в течение {time} часов");
    }

    // Перегруженный метод Drive с параметром bool
    public virtual void Drive(bool isEmergency)
    {
        if (isEmergency)
            Console.WriteLine($"{_brand} включил аварийный режим движения!");
        else
            Console.WriteLine($"{_brand} движется в обычном режиме");
    }

    // Виртуальный метод для отображения информации
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Транспортное средство: {_brand}, Год выпуска: {_year}");
    }
}

// Производный класс Car с переопределением методов
public class Car : Vehicle
{
    private int _doors;

    public Car(string brand, int year, int doors) : base(brand, year)
    {
        _doors = doors;
    }

    public int Doors
    {
        get { return _doors; }
        set 
        { 
            if (value >= 2 && value <= 5)
                _doors = value;
            else
                throw new ArgumentException("Некорректное количество дверей");
        }
    }

    // === ПЕРЕОПРЕДЕЛЕНИЕ МЕТОДОВ ===

    // Переопределение метода Drive с одним параметром
    public override void Drive(int speed)
    {
        if (speed > 120)
            Console.WriteLine($"{Brand} (автомобиль) превышает скорость! Текущая: {speed} км/ч");
        else
            Console.WriteLine($"{Brand} (автомобиль) едет со скоростью {speed} км/ч");
    }

    // Переопределение метода Drive с двумя параметрами
    public override void Drive(int speed, string direction)
    {
        // Вызов метода базового класса с помощью base
        base.Drive(speed, direction);
        Console.WriteLine($"Автомобиль аккуратно поворачивает {direction}");
    }

    // Переопределение метода Drive с параметром bool
    public override void Drive(bool isEmergency)
    {
        if (isEmergency)
        {
            Console.WriteLine($"{Brand} (автомобиль) включил аварийную сигнализацию!");
            Console.WriteLine("Все системы переведены в аварийный режим!");
        }
        else
        {
            base.Drive(isEmergency);
        }
    }

    // Переопределение метода DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип: Автомобиль, Количество дверей: {_doors}");
    }

    // Уникальный метод для Car
    public void Park(string parkingType)
    {
        Console.WriteLine($"{Brand} паркуется на {parkingType} парковке");
    }
}

// Производный класс Motorcycle с переопределением методов
public class Motorcycle : Vehicle
{
    private bool _hasSidecar;

    public Motorcycle(string brand, int year, bool hasSidecar) : base(brand, year)
    {
        _hasSidecar = hasSidecar;
    }

    public bool HasSidecar
    {
        get { return _hasSidecar; }
        set { _hasSidecar = value; }
    }

    // === ПЕРЕОПРЕДЕЛЕНИЕ МЕТОДОВ ===

    // Переопределение метода Drive с одним параметром
    public override void Drive(int speed)
    {
        if (speed > 80 && _hasSidecar)
            Console.WriteLine($"{Brand} (мотоцикл с коляской) осторожно едет {speed} км/ч");
        else if (speed > 100)
            Console.WriteLine($"{Brand} (мотоцикл) быстро мчится {speed} км/ч");
        else
            Console.WriteLine($"{Brand} (мотоцикл) едет со скоростью {speed} км/ч");
    }

    // Переопределение метода Drive с двумя параметрами
    public override void Drive(int speed, string direction)
    {
        Console.WriteLine($"{Brand} (мотоцикл) наклоняется в повороте {direction} на скорости {speed} км/ч");
    }

    // Переопределение метода Drive с параметром bool
    public override void Drive(bool isEmergency)
    {
        if (isEmergency)
        {
            Console.WriteLine($"{Brand} (мотоцикл) экстренно тормозит!");
            Console.WriteLine("ABS активирована!");
        }
        else
        {
            Console.WriteLine($"{Brand} (мотоцикл) наслаждается поездкой");
        }
    }

    // Переопределение метода DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        string sidecarInfo = _hasSidecar ? "с коляской" : "без коляски";
        Console.WriteLine($"Тип: Мотоцикл {sidecarInfo}");
    }

    // Уникальный метод для Motorcycle
    public void Wheelie()
    {
        Console.WriteLine($"{Brand} выполняет вилли (езда на заднем колесе)!");
    }
}

        Console.WriteLine("=== ПЕРЕГРУЗКА И ПЕРЕОПРЕДЕЛЕНИЕ МЕТОДОВ ===\n");

        // Создаем объекты разных типов
        Vehicle vehicle = new Vehicle("Generic Vehicle", 2020);
        Car car = new Car("Toyota", 2022, 4);
        Motorcycle motorcycle = new Motorcycle("Harley Davidson", 2021, false);
        Motorcycle motorcycleWithSidecar = new Motorcycle("Ural", 2020, true);

        // Демонстрация перегрузки методов
        Console.WriteLine("1. ПЕРЕГРУЗКА МЕТОДОВ (класс Vehicle):");
        Console.WriteLine(new string('-', 50));
        
        vehicle.Drive(60);
        vehicle.Drive(60, "на север");
        vehicle.Drive(60, "на восток", 2);
        vehicle.Drive(true);

        Console.WriteLine("\n2. ПЕРЕОПРЕДЕЛЕНИЕ МЕТОДОВ (разные классы):");
        Console.WriteLine(new string('-', 50));
        
        // Полиморфное поведение - один интерфейс, разные реализации
        Vehicle[] vehicles = { vehicle, car, motorcycle, motorcycleWithSidecar };
        
        foreach (var v in vehicles)
        {
            v.Drive(90); // Вызов переопределенного метода
            Console.WriteLine();
        }

        Console.WriteLine("3. ДЕМОНСТРАЦИЯ РАЗНЫХ ВАРИАНТОВ ПЕРЕГРУЗКИ:");
        Console.WriteLine(new string('-', 50));
        
        // Демонстрация разных перегруженных методов для Car
        car.Drive(150); // Один параметр
        car.Drive(80, "налево"); // Два параметра
        car.Drive(70, "прямо", 3); // Три параметра
        car.Drive(true); // Bool параметр

        Console.WriteLine("\n4. ВЫЗОВ МЕТОДОВ БАЗОВОГО КЛАССА (base):");
        Console.WriteLine(new string('-', 50));
        
        car.Drive(100, "направо"); // В этом методе используется base.Drive()

        Console.WriteLine("\n5. УНИКАЛЬНЫЕ МЕТОДЫ ПРОИЗВОДНЫХ КЛАССОВ:");
        Console.WriteLine(new string('-', 50));
        
        car.Park("подземной");
        motorcycle.Wheelie();

        Console.WriteLine("\n6. РАБОТА СО СВОЙСТВАМИ И ИНФОРМАЦИЯ:");
        Console.WriteLine(new string('-', 50));
        
        // Демонстрация работы свойств
        try
        {
            car.Year = 2023;
            car.Doors = 5;
            Console.WriteLine($"Автомобиль {car.Brand} {car.Year} года, {car.Doors} дверей");
            
            motorcycle.HasSidecar = true;
            Console.WriteLine($"Мотоцикл {motorcycle.Brand} теперь {(motorcycle.HasSidecar ? "с коляской" : "без коляски")}");
            
            // Попытка установить некорректные значения
            // car.Year = 1700; // Выбросит исключение
            // car.Doors = 6;   // Выбросит исключение
        }
        catch (ArgumentException ex)
        {
            Console.WriteLine($"Ошибка: {ex.Message}");
        }

        Console.WriteLine("\n7. ИНФОРМАЦИЯ О ВСЕХ ТРАНСПОРТНЫХ СРЕДСТВАХ:");
        Console.WriteLine(new string('-', 50));
        
        foreach (var v in vehicles)
        {
            v.DisplayInfo(); // Полиморфный вызов
            Console.WriteLine();
        }

        
        
        // Демонстрация разных реакций на одинаковые параметры
        Vehicle carAsVehicle = new Car("BMW", 2023, 2);
        Vehicle motorcycleAsVehicle = new Motorcycle("Honda", 2022, false);
        
        carAsVehicle.Drive(130); // Будет вызван переопределенный метод Car
        motorcycleAsVehicle.Drive(130); // Будет вызван переопределенный метод Motorcycle
````

````

// Базовый класс Animal
public class Animal
{
    // Поля с защищенным доступом
    protected string _name;
    protected int _age;
    protected double _weight;
    protected string _sound;

    // Конструктор базового класса
    public Animal(string name, int age, double weight, string sound = "издает звук")
    {
        _name = name;
        _age = age;
        _weight = weight;
        _sound = sound;
    }

    // Свойства с геттерами и сеттерами
    public string Name
    {
        get { return _name; }
        set 
        { 
            if (!string.IsNullOrWhiteSpace(value))
                _name = value;
            else
                throw new ArgumentException("Имя не может быть пустым");
        }
    }
    
    public int Age
    {
        get { return _age; }
        set 
        { 
            if (value >= 0)
                _age = value;
            else
                throw new ArgumentException("Возраст не может быть отрицательным");
        }
    }
    
    public double Weight
    {
        get { return _weight; }
        set 
        { 
            if (value > 0)
                _weight = value;
            else
                throw new ArgumentException("Вес должен быть положительным");
        }
    }

    public string Sound
    {
        get { return _sound; }
        set { _sound = value; }
    }

    // === ПЕРЕГРУЗКА МЕТОДОВ ===

    // Виртуальный метод MakeSound без параметров
    public virtual void MakeSound()
    {
        Console.WriteLine($"{_name} {_sound}");
    }

    // Перегруженный метод MakeSound с параметром (громкость)
    public virtual void MakeSound(int volume)
    {
        string volumeDescription = volume switch
        {
            > 80 => "очень громко",
            > 50 => "громко",
            > 20 => "нормально",
            _ => "тихо"
        };
        Console.WriteLine($"{_name} {_sound} {volumeDescription} (уровень {volume})");
    }

    // Перегруженный метод MakeSound с параметром (количество повторений)
    public virtual void MakeSound(string repetition)
    {
        if (repetition.ToLower() == "много")
            Console.WriteLine($"{_name} {_sound} много раз подряд!");
        else if (repetition.ToLower() == "один")
            Console.WriteLine($"{_name} {_sound} один раз");
        else
            Console.WriteLine($"{_name} {_sound} несколько раз");
    }

    // === ПЕРЕОПРЕДЕЛЕНИЕ (ПЕРЕКРЫТИЕ) МЕТОДОВ ===

    // Виртуальный метод Eat с параметрами
    public virtual void Eat(string food, int quantity)
    {
        Console.WriteLine($"{_name} съедает {quantity} единиц {food}");
        _weight += quantity * 0.1;
    }

    // Перегруженный метод Eat с одним параметром
    public virtual void Eat(string food)
    {
        Console.WriteLine($"{_name} ест {food}");
        _weight += 0.5;
    }

    // Виртуальный метод для получения информации
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Животное: {_name}, Возраст: {_age} лет, Вес: {_weight} кг");
    }

    // Виртуальный метод Sleep с перегрузкой
    public virtual void Sleep()
    {
        Console.WriteLine($"{_name} спит");
    }

    public virtual void Sleep(int hours)
    {
        Console.WriteLine($"{_name} спит {hours} часов");
        _weight -= hours * 0.01; // Немного теряет вес во время сна
    }
}

// Производный класс Dog
public class Dog : Animal
{
    private string _breed;
    private bool _isTrained;

    // Конструктор
    public Dog(string name, int age, double weight, string breed, bool isTrained = false) 
        : base(name, age, weight, "лает: Гав-гав!")
    {
        _breed = breed;
        _isTrained = isTrained;
    }

    // Свойства
    public string Breed
    {
        get { return _breed; }
        set { _breed = value; }
    }

    public bool IsTrained
    {
        get { return _isTrained; }
        set { _isTrained = value; }
    }

    // === ПЕРЕОПРЕДЕЛЕНИЕ МЕТОДОВ ===

    // Переопределение метода MakeSound
    public override void MakeSound()
    {
        string trainedInfo = _isTrained ? " (дрессированная)" : "";
        Console.WriteLine($"{_name} породы {_breed}{trainedInfo} {_sound}");
    }

    // Переопределение метода MakeSound с параметром громкости
    public override void MakeSound(int volume)
    {
        if (volume > 70)
            Console.WriteLine($"{_name} ГРОМКО ЛАЕТ: ГАВ-ГАВ-ГАВ! (уровень {volume})");
        else
            base.MakeSound(volume);
    }

    // Переопределение метода Eat
    public override void Eat(string food, int quantity)
    {
        if (food.ToLower() == "кость")
        {
            Console.WriteLine($"{_name} с радостью грызет {quantity} костей!");
            _weight += quantity * 0.2;
        }
        else if (food.ToLower() == "мясо")
        {
            Console.WriteLine($"{_name} быстро съедает {quantity} кусков мяса");
            _weight += quantity * 0.3;
        }
        else
        {
            base.Eat(food, quantity);
        }
    }

    // Переопределение метода DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        string trainedStatus = _isTrained ? "дрессированная" : "не дрессированная";
        Console.WriteLine($"  Порода: {_breed}, {trainedStatus}");
    }

    // Уникальный метод для Dog
    public void Fetch(string item)
    {
        string reaction = _isTrained ? "быстро приносит" : "медленно приносит";
        Console.WriteLine($"{_name} {reaction} {item}");
    }

    // Перегрузка метода Fetch
    public void Fetch(string item, int distance)
    {
        Console.WriteLine($"{_name} бежит за {item} на расстояние {distance} метров");
        Fetch(item);
    }
}

// Производный класс Cat
public class Cat : Animal
{
    private int _lives;
    private bool _isIndoor;

    // Конструктор
    public Cat(string name, int age, double weight, int lives = 9, bool isIndoor = true) 
        : base(name, age, weight, "мяукает: Мяу-мяу!")
    {
        _lives = lives;
        _isIndoor = isIndoor;
    }

    // Свойства
    public int Lives
    {
        get { return _lives; }
        set 
        { 
            if (value >= 0 && value <= 9)
                _lives = value;
            else
                throw new ArgumentException("Количество жизней должно быть от 0 до 9");
        }
    }

    public bool IsIndoor
    {
        get { return _isIndoor; }
        set { _isIndoor = value; }
    }

    // === ПЕРЕОПРЕДЕЛЕНИЕ МЕТОДОВ ===

    // Переопределение метода MakeSound
    public override void MakeSound()
    {
        string location = _isIndoor ? "домашний" : "уличный";
        Console.WriteLine($"{_name} ({location} кот) {_sound}");
    }

    // Переопределение метода MakeSound с параметром повторений
    public override void MakeSound(string repetition)
    {
        if (repetition.ToLower() == "много")
            Console.WriteLine($"{_name} настойчиво мяукает много раз: Мяу! Мяу! Мяу!");
        else
            base.MakeSound(repetition);
    }

    // Переопределение метода Eat
    public override void Eat(string food, int quantity)
    {
        if (food.ToLower() == "рыба")
        {
            Console.WriteLine($"{_name} с удовольствием съедает {quantity} рыб!");
            _weight += quantity * 0.15;
        }
        else if (food.ToLower() == "молоко")
        {
            Console.WriteLine($"{_name} лакает {quantity} мисок молока");
            _weight += quantity * 0.05;
        }
        else
        {
            Console.WriteLine($"{_name} привередливо обнюхивает {food}");
            base.Eat(food, quantity);
        }
    }

    // Переопределение метода Sleep
    public override void Sleep(int hours)
    {
        Console.WriteLine($"{_name} мурлыкая спит {hours} часов на диване");
        _weight -= hours * 0.005; // Коты меньше теряют вес
    }

    // Переопределение метода DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        string location = _isIndoor ? "домашний" : "уличный";
        Console.WriteLine($"  Тип: {location} кот, Осталось жизней: {_lives}");
    }

    // Уникальный метод для Cat
    public void Climb(string objectName)
    {
        Console.WriteLine($"{_name} ловко забирается на {objectName}");
    }

    // Перегрузка метода Climb
    public void Climb(string objectName, int height)
    {
        Console.WriteLine($"{_name} забирается на {objectName} высотой {height} метров");
    }
}

// Производный класс Bird
public class Bird : Animal
{
    private double _wingspan;
    private bool _canFly;

    // Конструктор
    public Bird(string name, int age, double weight, double wingspan, bool canFly = true) 
        : base(name, age, weight, "чирикает: Чик-чирик!")
    {
        _wingspan = wingspan;
        _canFly = canFly;
    }

    // Свойства
    public double Wingspan
    {
        get { return _wingspan; }
        set 
        { 
            if (value > 0)
                _wingspan = value;
            else
                throw new ArgumentException("Размах крыльев должен быть положительным");
        }
    }

    public bool CanFly
    {
        get { return _canFly; }
        set { _canFly = value; }
    }

    // === ПЕРЕОПРЕДЕЛЕНИЕ МЕТОДОВ ===

    // Переопределение метода MakeSound
    public override void MakeSound()
    {
        string flyAbility = _canFly ? "летающая" : "нелетающая";
        Console.WriteLine($"{_name} ({flyAbility} птица) {_sound}");
    }

    // Переопределение метода MakeSound с параметром громкости
    public override void MakeSound(int volume)
    {
        if (volume > 60)
            Console.WriteLine($"{_name} ГРОМКО ЩЕБЕТАЕТ: ЧИК-ЧИРИК-ЧИК! (уровень {volume})");
        else if (volume < 20)
            Console.WriteLine($"{_name} тихо щебечет... чик... чик... (уровень {volume})");
        else
            base.MakeSound(volume);
    }

    // Переопределение метода Eat
    public override void Eat(string food)
    {
        if (food.ToLower() == "зерно")
        {
            Console.WriteLine($"{_name} клюет {food}");
            _weight += 0.1;
        }
        else if (food.ToLower() == "насекомые")
        {
            Console.WriteLine($"{_name} ловит {food}");
            _weight += 0.05;
        }
        else
        {
            base.Eat(food);
        }
    }

    public override void Eat(string food, int quantity)
    {
        if (food.ToLower() == "зерно")
        {
            Console.WriteLine($"{_name} клюет {quantity} зернышек");
            _weight += quantity * 0.02;
        }
        else if (food.ToLower() == "насекомые")
        {
            Console.WriteLine($"{_name} ловит {quantity} насекомых");
            _weight += quantity * 0.01;
        }
        else
        {
            Console.WriteLine($"{_name} осторожно пробует {food}");
            _weight += quantity * 0.01;
        }
    }

    // Переопределение метода DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        string flightAbility = _canFly ? "летающая" : "нелетающая";
        Console.WriteLine($"  Тип: {flightAbility} птица, Размах крыльев: {_wingspan} см");
    }

    // Уникальный метод для Bird
    public void Fly(int distance)
    {
        if (_canFly)
            Console.WriteLine($"{_name} пролетает {distance} метров в небе!");
        else
            Console.WriteLine($"{_name} пытается взлететь, но не может летать");
    }

    // Перегрузка метода Fly
    public void Fly(int distance, int height)
    {
        if (_canFly)
            Console.WriteLine($"{_name} пролетает {distance} метров на высоте {height} метров!");
        else
            Console.WriteLine($"{_name} безуспешно пытается взлететь");
    }
}

// Производный класс Fish
public class Fish : Animal
{
    private string _waterType;
    private double _swimSpeed;

    // Конструктор
    public Fish(string name, int age, double weight, string waterType = "пресная", double swimSpeed = 5.0) 
        : base(name, age, weight, "пускает пузыри: Буль-буль!")
    {
        _waterType = waterType;
        _swimSpeed = swimSpeed;
    }

    // Свойства
    public string WaterType
    {
        get { return _waterType; }
        set { _waterType = value; }
    }

    public double SwimSpeed
    {
        get { return _swimSpeed; }
        set 
        { 
            if (value >= 0)
                _swimSpeed = value;
            else
                throw new ArgumentException("Скорость плавания не может быть отрицательной");
        }
    }

    // === ПЕРЕОПРЕДЕЛЕНИЕ МЕТОДОВ ===

    // Переопределение метода MakeSound
    public override void MakeSound()
    {
        Console.WriteLine($"{_name} ({_waterType} рыба) {_sound}");
    }

    // Переопределение метода MakeSound с параметром повторений
    public override void MakeSound(string repetition)
    {
        if (repetition.ToLower() == "много")
            Console.WriteLine($"{_name} активно пускает пузыри: Буль-буль-буль-буль!");
        else
            base.MakeSound(repetition);
    }

    // Переопределение метода Eat
    public override void Eat(string food, int quantity)
    {
        if (food.ToLower() == "планктон")
        {
            Console.WriteLine($"{_name} фильтрует {quantity} планктона из воды");
            _weight += quantity * 0.001;
        }
        else if (food.ToLower() == "водоросли")
        {
            Console.WriteLine($"{_name} nibbles on {quantity} водорослей");
            _weight += quantity * 0.005;
        }
        else
        {
            Console.WriteLine($"{_name} игнорирует {food} и ищет другую еду");
        }
    }

    // Переопределение метода Sleep
    public override void Sleep()
    {
        Console.WriteLine($"{_name} плавает во сне с открытыми глазами");
    }

    // Переопределение метода DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Тип: {_waterType} рыба, Скорость плавания: {_swimSpeed} км/ч");
    }

    // Уникальный метод для Fish
    public void Swim(int depth)
    {
        Console.WriteLine($"{_name} плавает на глубине {depth} метров");
    }

    // Перегрузка метода Swim
    public void Swim(int depth, string direction)
    {
        Console.WriteLine($"{_name} плывет {direction} на глубине {depth} метров");
    }
}

        // Создаем список животных разных типов
        List<Animal> animals = new List<Animal>
        {
            new Dog("Бобик", 3, 12.5, "Овчарка", true),
            new Cat("Мурзик", 2, 4.2, 7, true),
            new Bird("Кеша", 1, 0.3, 25.5, true),
            new Fish("Немо", 1, 0.5, "соленая", 8.0),
            new Bird("Страус", 4, 45.0, 200.0, false),
            new Dog("Шарик", 5, 8.7, "Дворняжка", false)
        };

        // Демонстрация перегрузки методов
        Console.WriteLine("1. ПЕРЕГРУЗКА МЕТОДА MakeSound:");
        Console.WriteLine(new string('-', 50));
        
        Animal testAnimal = new Animal("Тестовое животное", 1, 10.0);
        testAnimal.MakeSound();              // Без параметров
        testAnimal.MakeSound(75);            // С параметром int
        testAnimal.MakeSound("много");       // С параметром string

        Console.WriteLine("\n2. ПЕРЕОПРЕДЕЛЕНИЕ МЕТОДОВ (полиморфизм):");
        Console.WriteLine(new string('-', 50));
        
        // Демонстрация полиморфного поведения
        foreach (var animal in animals)
        {
            animal.MakeSound(); // Вызов переопределенного метода
        }

        Console.WriteLine("\n3. ПЕРЕГРУЗКА МЕТОДА MakeSound В ПРОИЗВОДНЫХ КЛАССАХ:");
        Console.WriteLine(new string('-', 50));
        
        animals[0].MakeSound(85);  // Собака громко лает
        animals[1].MakeSound("много"); // Кот много мяукает
        animals[2].MakeSound(15);  // Птица тихо щебечет

        Console.WriteLine("\n4. ПЕРЕГРУЗКА И ПЕРЕОПРЕДЕЛЕНИЕ МЕТОДА Eat:");
        Console.WriteLine(new string('-', 50));
        
        // Демонстрация разных вариантов метода Eat
        animals[0].Eat("кость", 2);        // Собака
        animals[0].Eat("кость");           // Перегруженный метод
        animals[1].Eat("рыба", 1);         // Кот
        animals[2].Eat("зерно");           // Птица (перегруженный)
        animals[3].Eat("планктон", 100);   // Рыба

        Console.WriteLine("\n5. ИНФОРМАЦИЯ О ВСЕХ ЖИВОТНЫХ:");
        Console.WriteLine(new string('-', 50));
        
        foreach (var animal in animals)
        {
            animal.DisplayInfo();
            Console.WriteLine();
        }

        Console.WriteLine("6. УНИКАЛЬНЫЕ МЕТОДЫ И ИХ ПЕРЕГРУЗКА:");
        Console.WriteLine(new string('-', 50));
        
        // Демонстрация уникальных методов через приведение типов
        foreach (var animal in animals)
        {
            if (animal is Dog dog)
            {
                dog.Fetch("палку");
                dog.Fetch("мяч", 20); // Перегруженный метод
            }
            else if (animal is Cat cat)
            {
                cat.Climb("дерево");
                cat.Climb("забор", 2); // Перегруженный метод
            }
            else if (animal is Bird bird)
            {
                bird.Fly(50);
                bird.Fly(100, 10); // Перегруженный метод
            }
            else if (animal is Fish fish)
            {
                fish.Swim(5);
                fish.Swim(3, "вперед"); // Перегруженный метод
            }
            Console.WriteLine();
        }

        Console.WriteLine("7. РАБОТА СО СВОЙСТВАМИ:");
        Console.WriteLine(new string('-', 50));
        
        // Демонстрация работы свойств с валидацией
        try
        {
            Dog myDog = new Dog("Рекс", 3, 15.0, "Лабрадор", true);
            myDog.Name = "Новый Рекс";
            myDog.Age = 4;
            myDog.Weight = 16.5;
            myDog.Breed = "Золотистый ретривер";
            Console.WriteLine($"Собака: {myDog.Name}, {myDog.Age} лет, {myDog.Weight} кг, {myDog.Breed}");

            // Попытка установить некорректные значения
            // myDog.Age = -1; // Выбросит исключение
            // myDog.Weight = 0; // Выбросит исключение
        }
        catch (ArgumentException ex)
        {
            Console.WriteLine($"Ошибка: {ex.Message}");
        }

        Console.WriteLine("\n8. ДОПОЛНИТЕЛЬНАЯ ДЕМОНСТРАЦИЯ ПЕРЕОПРЕДЕЛЕНИЯ Sleep:");
        Console.WriteLine(new string('-', 50));
        
        animals[1].Sleep();    // Кот спит
        animals[1].Sleep(8);   // Кот спит 8 часов (перегруженный)
        animals[3].Sleep();    // Рыба спит (переопределенный)
````
